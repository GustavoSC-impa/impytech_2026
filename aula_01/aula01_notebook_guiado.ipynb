> ## **Conceito 1 - Exibição de informações e variáveis**
Vamos aprender a criar os primeiros códigos utilizando a linguagem **Python**.

> **O que vamos aprender:**
> - Exibir informação na tela com `print()`
> - Criar variáveis
> - Receber informações utilizando `input()`
> - Tipos de variáveis: `str`, `int` e `float`

### **Passo 1.1 — Criando e exibindo variáveis**
**Variáveis** são como espaços alocados na memória do computador para guardar informações. Para **exibir informação**, usamos a função `print()`, e para **receber informação**, usamos a função `input()`. <br><br>
Além disso, temos três principais **tipos de variáveis** que estuaremos inicialmente:
| Tipo               | Nomenclatura | Descrição                                                        |
|--------------------|--------------|------------------------------------------------------------------|
| Texto              | `str`        | Sempre deve ser declarada **entre aspas** (simples ou duplas).   |
| Números inteiros   | `int`        | Números **sem casas decimais**.                                  |
| Números decimais   | `float`      | Sempre deve usar **ponto** ao invés de vírgula.                  |

```python
# =============================================
# PASSO 1.1 — Criando e exibindo variáveis
# =============================================

# [PARTE 1] — Criando variáveis com diferentes valores
nome = "Ana"          # variável do tipo texto (str)
idade = 20            # variável do tipo inteiro (int)
altura = 1.65         # variável do tipo decimal (float)

# [PARTE 2] — Exibindo os valores com print()
print(nome)           # exibe: Ana
print(idade)          # exibe: 20
print(altura)         # exibe: 1.65
```

> 📝 Explicação do código acima:
> - `nome = "Ana"` → O sinal `=` **atribui** o valor `"Ana"` à variável `nome`. Como é um texto, sempre devemos colocar entre aspas (`"` ou `'`)
> - `idade = 20` → Atribui o valor `20` a variável `idade`.
> - `altura = 1.65` → Atribui o valor `1.65` a variável `altura`. Como é um número com casas decimais, precisamos **sempre** utilizar ponto ao invés de vírgula.
> - **Ponto de atenção:** Python diferencia maiúsculas de minúsculas — `Nome` e `nome` são variáveis diferentes!

### **Passo 1.2 — Recebendo valores**
A função `input()` serve para recebermos uma informação digitada. Para isso, precisamos guardar o que foi digitado em alguma **variável**.

```python
# =============================================
# PASSO 1.2 — Recebendo valores
# =============================================

# [PARTE 1] — Recebendo o nome do usuário
seu_nome = str(input("Qual é o seu nome? "))  # A variável "seu_nome" terá o valor que você escrever

# [PARTE 2] — Exibindo o valor
print(seu_nome)
```

> 📝 Explicação do código acima:
> - `seu_nome = str( ... )` → Indica que a variável `seu_nome` terá tipo `string(str)`, que é usado para texto
> - `seu_nome = str(input("..."))` → O método `input()` pede para o usuário digitar algo
> - `seu_nome = str(input("Qual é o seu nome? "))` → O programa irá exibir o texto `Qual é o seu nome? ` na caixa que será aberta para escrevermos
> - `print(seu_nome)` → Exibe o que o usuário digitou

> ## **Conceito 2 - Operadores aritméticos**
Quando estamos trabalhando com variáveis do tipo `int` e `float`, podemos realizar diversos cálculos aritméticos com estes valores.

> **O que vamos aprender:**

### **Passo 2.1 — Os principais operadores aritméticos**
Os principais **operadores aritméticos** que podemos usar entre dois **números** no Python são:

| Operação         | Operador |
|------------------|----------|
| Adição           | `+`      |
| Subtração        | `-`      |
| Multiplicação    | `*`      |
| Divisão          | `/`      |
| Potência         | `**`     |
| Divisão inteira  | `//`     |
| Módulo           | `%`      |

```python
# =============================================
# PASSO 2.1 — Os principais operadores aritméticos
# =============================================

# [PARTE 1] — Um exemplo de cada tipo
print(5 + 5)    # Usa o operador de SOMA para calcular 5 + 5
print(3 - 2)    # Usa o operador de SUBTRAÇÃO para calcular 3 - 2
print(6 * 2)    # Usa o operador de MULTIPLICAÇÃO para calcular 6 × 2
print(10 / 2)   # Usa o operador de DIVISÃO para calcular 10 ÷ 2
print(2 ** 3)   # Usa o operador de POTÊNCIA para calcular 2³
print(5 // 2)   # Usa o operador de DIVISÃO INTEIRA (descarta a parte decimal)
print(5 % 2)    # Usa o operador de MÓDULO para obter o resto da divisão de 5 por 2

variavel = 10 + 10  # Experimente trocar o operador e os números
print(variavel)
```

> 📝 Explicação do código acima:
> - `5 + 5` → Usa o operador `+` para realizar uma **adição** entre os números `5` e `5`.
> - `3 - 2` → Usa o operador `-` para realizar uma **subtração**, calculando `3 - 2`.
> - `6 * 2` → Usa o operador `*` para realizar uma **multiplicação**, calculando `6 × 2`.
> - `10 / 2` → Usa o operador `/` para realizar uma **divisão**, retornando um número decimal (`float`), mesmo que o resultado seja inteiro.
> - `2 ** 3` → Usa o operador `**` para calcular uma **potência**, ou seja, `2³`.
> - `5 // 2` → Usa o operador `//` para realizar uma **divisão inteira**, descartando a parte decimal do resultado.
> - `5 % 2` → Usa o operador `%` para calcular o **resto da divisão** de `5` por `2`.


### **Passo 2.2 - Expressões aritméticas**
Também podemos misturar operadores e números para criar **expressões aritméticas**


**Temperatura média (versão 1)**

```python
# =============================================
# PASSO 2.2 — Expressões aritméticas
# =============================================

# [PARTE 1] — Temperatura média estática
temperatura_1 = 37.91
temperatura_2 = 37.12
temperatura_3 = 38.01

soma = temperatura_1 + temperatura_2 + temperatura_3
media = soma/3

print("A temperatura média do paciente é:")
print(media)
```

**Temperatura média (versão 2)**



```python
temperatura_1 = float(input("Digite a temperatura 1: "))
temperatura_2 = float(input("Digite a temperatura 2: "))
temperatura_3 = float(input("Digite a temperatura 3: "))

soma = temperatura_1 + temperatura_2 + temperatura_3
media = soma/3

print("A temperatura média do paciente é:")
print(media)
```

**Temperatura média (versão 3)**

```python
media = (37.91 + 37.12 + 38.01)/3

print("A temperatura média do paciente é:")
print(media)
```

### **Passo 2.3 - Ordem de precedência**
Assim como na matemática, as operações de uma expressão aritméticas obedecem uma ordem de precedência para resoluções (Ex.: multiplicações são feitas primeiros que somas)

No Python, a ordem de precedência das operações são:

| Operador(es) | Prioridade |
| :---: | :---: |
| `( )` | 1º |
| `**` | 2º |
| `*` `/` `//` `%` | 3º |
| `+` `-` | 4º |



```python
# =============================================
# PASSO 2.3 — Ordem de precedência
# =============================================

# [PARTE 1] — Testando precedências
media_1 = 4+4/2   # Sem usar parênteses (média errada)
media_2 = (4+4)/2 # Usando parêntes (média correta)

print(media_1)
print(media_2)
```

> 📝 Explicação do código acima:
> - `media_1 = 4+4/2` → Pela tabela de ordem de precedência, o operador de divisão `/` tem mais prioridade que o operador de soma `+`, então a **divisão será feita primeiro**. Então, a expressão equivalente é: <br>
`4+4/2 = 4+2 = 6`

> - `media_2 = (4+4)/2` → Pela tabela de ordem de precedência, os parênteses `()` tem mais prioridade que a divisão `/`, então **a expressão dos parênteses será feita primeiro**. Como dentro dos parênteses só temos uma soma, a expressão equivalente é: <br>
`(4+4)/2 = 8/2 = 4`
