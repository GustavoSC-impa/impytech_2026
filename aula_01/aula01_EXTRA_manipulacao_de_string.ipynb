> ## **Conceito 1 — Descobrindo o tamanho de um texto**
Uma **string** é uma sequência de caracteres. A função `len()` informa quantos caracteres existem em um texto.

> **O que vamos aprender:**
> - Usar `len()` para descobrir o tamanho de uma string
> - Entender que espaços e sinais de pontuação também são caracteres

### **Passo 1.1 — Usando a função `len()`**
Vamos contar quantos caracteres existem em diferentes textos.

```python
# =============================================
# PASSO 1.1 — Descobrindo o tamanho de uma string
# =============================================

# [PARTE 1] — Contando os caracteres de uma palavra
palavra = "Python"

print(len(palavra))  # exibe: 6

# [PARTE 2] — Espaços e pontuação também são contados
frase = "Hello World!"

print(len(frase))    # exibe: 12

# AGORA É SUA VEZ:
# Crie uma variável com uma frase de sua escolha e exiba seu tamanho.
```

> 📝 Explicação do código acima:
> - `len(texto)` → devolve a quantidade de caracteres da string
> - Em `"Hello World!"`, o espaço e o ponto de exclamação também são contados
> - **Ponto de atenção:** `len()` conta caracteres, não palavras

> ## **Conceito 2 — Acessando caracteres de uma string**
Cada caractere de uma string possui uma posição chamada **índice**. Em Python, a contagem começa em `0`.

> **O que vamos aprender:**
> - Acessar caracteres por índices positivos
> - Usar índices negativos para contar a partir do final

### **Passo 2.1 — Índices positivos**
O primeiro caractere possui índice `0`, o segundo possui índice `1` e assim por diante.

```python
# =============================================
# PASSO 2.1 — Acessando caracteres por índice
# =============================================

palavra = "Python"

print(palavra[0])  # exibe: P
print(palavra[1])  # exibe: y
print(palavra[2])  # exibe: t

# AGORA É SUA VEZ:
# Exiba o quarto caractere da palavra.
```

> 📝 Explicação do código acima:
> - `palavra[0]` → acessa o primeiro caractere
> - `palavra[1]` → acessa o segundo caractere
> - **Ponto de atenção:** acessar uma posição inexistente causa `IndexError`

### **Passo 2.2 — Índices negativos**
Índices negativos permitem acessar os caracteres começando pelo final da string.

```python
# =============================================
# PASSO 2.2 — Usando índices negativos
# =============================================

palavra = "Python"

print(palavra[-1])  # último caractere: n
print(palavra[-2])  # penúltimo caractere: o

# AGORA É SUA VEZ:
# Exiba o third caractere contado a partir do final.
```

> 📝 Explicação do código acima:
> - `palavra[-1]` → acessa o último caractere
> - `palavra[-2]` → acessa o penúltimo caractere
> - Índices negativos são úteis quando não sabemos previamente o tamanho do texto

> ## **Conceito 3 — Recortando partes de um texto**
O **fatiamento** permite selecionar uma parte de uma string usando a sintaxe `texto[inicio:fim]`.

> **O que vamos aprender:**
> - Selecionar intervalos de caracteres
> - Omitir o início ou o final de uma fatia
> - Inverter uma string

### **Passo 3.1 — Fatiamento com início e fim**
O índice inicial é incluído, mas o índice final não é incluído.

```python
# =============================================
# PASSO 3.1 — Fatiamento de strings
# =============================================

palavra = "Programação"

print(palavra[0:4])  # exibe: Prog

# AGORA É SUA VEZ:
# Recorte e exiba os caracteres das posições 4 até 7.
```

> 📝 Explicação do código acima:
> - `palavra[0:4]` → seleciona os índices `0`, `1`, `2` e `3`
> - O índice `4` marca o final da fatia, mas não entra no resultado
> - Essa regra facilita dividir uma string em partes consecutivas

### **Passo 3.2 — Omitindo posições e invertendo o texto**
Podemos omitir um dos limites ou usar um passo negativo para percorrer o texto ao contrário.

```python
# =============================================
# PASSO 3.2 — Variações do fatiamento
# =============================================

palavra = "Programação"

print(palavra[:4])    # do início até o índice 4
print(palavra[4:])    # do índice 4 até o final
print(palavra[::-1])  # texto invertido

# AGORA É SUA VEZ:
# Exiba apenas os cinco últimos caracteres da palavra.
```

> 📝 Explicação do código acima:
> - `palavra[:4]` → começa no início da string
> - `palavra[4:]` → continua até o final da string
> - `palavra[::-1]` → percorre toda a string com passo `-1`, invertendo-a

> ## **Conceito 4 — Transformando letras maiúsculas e minúsculas**
Strings possuem métodos que criam versões modificadas do texto.

> **O que vamos aprender:**
> - Usar `.upper()` e `.lower()`
> - Usar `.title()` e `.capitalize()`
> - Entender que strings são imutáveis

### **Passo 4.1 — Convertendo todas as letras**
Os métodos `.upper()` e `.lower()` alteram a capitalização de todas as letras.

```python
# =============================================
# PASSO 4.1 — upper() e lower()
# =============================================

nome = "RYAN Kevin"

print(nome.upper())  # todas as letras maiúsculas
print(nome.lower())  # todas as letras minúsculas

# AGORA É SUA VEZ:
# Crie uma frase e exiba uma versão totalmente maiúscula.
```

> 📝 Explicação do código acima:
> - `texto.upper()` → devolve uma nova string em letras maiúsculas
> - `texto.lower()` → devolve uma nova string em letras minúsculas
> - Esses métodos são úteis para padronizar textos antes de compará-los

### **Passo 4.2 — Capitalizando palavras**
Os métodos `.title()` e `.capitalize()` produzem resultados diferentes.

```python
# =============================================
# PASSO 4.2 — title() e capitalize()
# =============================================

nome = "ryan kevin felinto"
frase = "python é uma linguagem de programação"

print(nome.title())
print(frase.capitalize())

# AGORA É SUA VEZ:
# Teste os dois métodos usando a mesma frase e compare os resultados.
```

> 📝 Explicação do código acima:
> - `texto.title()` → coloca a primeira letra de cada palavra em maiúscula
> - `texto.capitalize()` → coloca apenas a primeira letra do texto em maiúscula
> - **Ponto de atenção:** os métodos também podem modificar a capitalização das demais letras

### **Passo 4.3 — Strings são imutáveis**
Os métodos não alteram a string original. Eles devolvem uma nova string.

```python
# =============================================
# PASSO 4.3 — Imutabilidade das strings
# =============================================

nome = "ryan"
nome_maiusculo = nome.upper()

print(nome)            # continua sendo: ryan
print(nome_maiusculo)  # nova string: RYAN

# AGORA É SUA VEZ:
# Crie uma nova variável contendo 'nome' com a primeira letra maiúscula.
```

> 📝 Explicação do código acima:
> - `nome.upper()` não altera o conteúdo guardado em `nome`
> - Para guardar o resultado, atribuímos a nova string a outra variável
> - Strings são chamadas de **imutáveis** porque seus caracteres não são alterados diretamente

> ## **Conceito 5 — Removendo espaços desnecessários**
Dados digitados pelo usuário podem conter espaços extras no início ou no final.

> **O que vamos aprender:**
> - Usar `.strip()`, `.lstrip()` e `.rstrip()`
> - Combinar `.strip()` com `input()`

### **Passo 5.1 — Limpando espaços**
O método `.strip()` remove espaços das duas extremidades da string.

```python
# =============================================
# PASSO 5.1 — Removendo espaços
# =============================================

nome = "   Ryan Kevin   "

print(nome.strip())   # remove dos dois lados
print(nome.lstrip())  # remove apenas do lado esquerdo
print(nome.rstrip())  # remove apenas do lado direito

# [PARTE 2] — Uso comum com input()
nome_usuario = input("Digite seu nome: ").strip()

print("Olá,", nome_usuario)

# AGORA É SUA VEZ:
# Peça uma cidade ao usuário, remova os espaços extras e exiba o resultado.
```

> 📝 Explicação do código acima:
> - `strip()` → remove espaços no início e no final
> - `lstrip()` → remove espaços apenas à esquerda
> - `rstrip()` → remove espaços apenas à direita
> - **Ponto de atenção:** espaços existentes no meio do texto não são removidos

> ## **Conceito 6 — Substituindo partes de uma string**
O método `.replace()` cria uma nova string substituindo um trecho por outro.

> **O que vamos aprender:**
> - Substituir palavras ou caracteres
> - Remover caracteres usando uma string vazia

### **Passo 6.1 — Usando `.replace()`**
Vamos substituir uma linguagem de programação e limpar um número de telefone.

```python
# =============================================
# PASSO 6.1 — Substituindo e removendo caracteres
# =============================================

# [PARTE 1] — Substituindo um trecho
frase = "Eu gosto de C++"
nova_frase = frase.replace("C++", "Python")

print(nova_frase)

# [PARTE 2] — Removendo caracteres
telefone = "(21) 99999-8888"

telefone_limpo = telefone.replace("(", "")
telefone_limpo = telefone_limpo.replace(")", "")
telefone_limpo = telefone_limpo.replace(" ", "")
telefone_limpo = telefone_limpo.replace("-", "")

print(telefone_limpo)

# AGORA É SUA VEZ:
# Troque a palavra "difícil" por "interessante" na frase:
# "Python é difícil".
```

> 📝 Explicação do código acima:
> - `texto.replace(antigo, novo)` → substitui todas as ocorrências encontradas
> - Usar `""` como novo texto remove o trecho escolhido
> - Assim como os outros métodos, `.replace()` devolve uma nova string

> ## **Conceito 7 — Separando uma string**
O método `.split()` divide uma string em várias partes e devolve uma lista.

> **O que vamos aprender:**
> - Separar um texto pelos espaços
> - Escolher um separador específico
> - Guardar cada parte em uma variável

### **Passo 7.1 — Dividindo textos**
Sem argumentos, `.split()` separa o texto nos espaços.

```python
# =============================================
# PASSO 7.1 — Separando strings com split()
# =============================================

# [PARTE 1] — Separando pelos espaços
frase = "Python é fácil de aprender"
palavras = frase.split()

print(palavras)

# [PARTE 2] — Escolhendo um separador
data = "01/07/2026"
partes = data.split("/")

print(partes)

# AGORA É SUA VEZ:
# Separe a string "azul-verde-amarelo" usando "-" como separador.
```

> 📝 Explicação do código acima:
> - `texto.split()` → separa o texto nos espaços
> - `texto.split("/")` → separa o texto sempre que encontra `/`
> - O resultado é uma **lista de strings**

### **Passo 7.2 — Guardando as partes em variáveis**
Quando sabemos quantas partes serão produzidas, podemos distribuí-las em variáveis.

```python
# =============================================
# PASSO 7.2 — Desempacotando o resultado de split()
# =============================================

data = "10/07/2026"

dia, mes, ano = data.split("/")

print("Dia:", dia)
print("Mês:", mes)
print("Ano:", ano)

# AGORA É SUA VEZ:
# Separe "Lucas-Silva" e guarde cada nome em uma variável.
```

> 📝 Explicação do código acima:
> - `data.split("/")` produz três partes
> - As três partes são atribuídas, na mesma ordem, a `dia`, `mes` e `ano`
> - **Ponto de atenção:** a quantidade de variáveis deve coincidir com a quantidade de partes

> ## **Conceito 8 — Juntando vários textos**
O método `.join()` faz o caminho contrário de `.split()`: ele combina várias strings usando um separador.

> **O que vamos aprender:**
> - Juntar palavras com espaços
> - Reconstruir textos usando outros separadores

### **Passo 8.1 — Usando `.join()`**
O separador aparece antes de `.join()`.

```python
# =============================================
# PASSO 8.1 — Juntando strings com join()
# =============================================

# [PARTE 1] — Juntando palavras
palavras = ["Python", "é", "muito", "útil"]
frase = " ".join(palavras)

print(frase)

# [PARTE 2] — Juntando partes de uma data
partes = ["10", "07", "2026"]
data = "/".join(partes)

print(data)

# AGORA É SUA VEZ:
# Junte ["2026", "07", "10"] usando "-" como separador.
```

> 📝 Explicação do código acima:
> - `" ".join(palavras)` → usa um espaço entre os elementos
> - `"/".join(partes)` → usa `/` entre os elementos
> - **Ponto de atenção:** todos os elementos precisam ser strings

> ## **Conceito 9 — Inserindo variáveis dentro de textos**
As **f-strings** permitem inserir variáveis e expressões diretamente dentro de uma mensagem.

> **O que vamos aprender:**
> - Criar mensagens com f-strings
> - Inserir expressões entre chaves
> - Controlar a quantidade de casas decimais

### **Passo 9.1 — Variáveis e expressões em f-strings**
Colocamos a letra `f` antes das aspas e usamos `{}` para inserir valores.

```python
# =============================================
# PASSO 9.1 — Criando mensagens com f-strings
# =============================================

# [PARTE 1] — Inserindo variáveis
nome = "Ryan"
idade = 19

mensagem = f"Meu nome é {nome} e tenho {idade} anos."
print(mensagem)

# [PARTE 2] — Inserindo uma expressão
preco = 25
quantidade = 4

print(f"O valor total é R$ {preco * quantidade}.")

# AGORA É SUA VEZ:
# Crie uma f-string que mostre seu nome e o ano em que nasceu.
```

> 📝 Explicação do código acima:
> - `f"texto {variavel}"` → insere o valor da variável na mensagem
> - Dentro de `{}`, também podemos escrever expressões
> - O resultado da expressão é calculado antes de ser inserido no texto

### **Passo 9.2 — Formatando casas decimais**
A notação `:.2f` exibe um número com duas casas decimais.

```python
# =============================================
# PASSO 9.2 — Formatando números decimais
# =============================================

preco_1 = 19.9
preco_2 = 19.879999

print(f"Preço 1: R$ {preco_1:.2f}")
print(f"Preço 2: R$ {preco_2:.2f}")

# AGORA É SUA VEZ:
# Exiba o número 3.14159265 com três casas decimais.
```

> 📝 Explicação do código acima:
> - `{preco:.2f}` → formata o número com exatamente duas casas decimais
> - O `f` indica formatação de número de ponto flutuante
> - O valor exibido é arredondado para a quantidade de casas solicitada

> ## **Conceito 10 — Exemplo completo: cadastro simples**
Agora vamos combinar vários recursos estudados em um único programa.

> **Recursos utilizados:**
> - `input()`
> - `.strip()`, `.title()` e `.lower()`
> - `len()`
> - f-strings
> - operador `in`
> - estrutura condicional `if`

### **Passo 10.1 — Construindo o cadastro**
O programa recebe, padroniza e exibe os dados informados pelo usuário.

```python
# =============================================
# PASSO 10.1 — Cadastro simples
# =============================================

# [PARTE 1] — Recebendo e padronizando os dados
nome = input("Digite seu nome completo: ").strip().title()
email = input("Digite seu e-mail: ").strip().lower()

# [PARTE 2] — Exibindo os dados cadastrados
print()
print("Dados cadastrados:")
print(f"Nome: {nome}")
print(f"E-mail: {email}")
print(f"Quantidade de caracteres do nome: {len(nome)}")

# [PARTE 3] — Verificação simples do formato do e-mail
if "@" in email and "." in email:
    print("O e-mail possui um formato válido.")
else:
    print("O e-mail pode estar incorreto.")

# AGORA É SUA VEZ:
# Acrescente uma entrada para a cidade e padronize-a com strip() e title().
```

> 📝 Explicação do código acima:
> - Os métodos podem ser encadeados: `input(...).strip().title()`
> - O nome é padronizado com `.title()` e o e-mail com `.lower()`
> - `len(nome)` conta os caracteres do nome já tratado
> - `"@" in email` verifica se o caractere `@` aparece na string
> - **Ponto de atenção:** essa verificação de e-mail é apenas básica e não garante que o endereço realmente exista
