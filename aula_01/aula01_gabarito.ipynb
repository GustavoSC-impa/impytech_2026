> ## **Problema 3 — Problemas com operações aritméticas** 🟡

Aplicando expressões matemáticas para resolver problemas simples.


### ✅ **Problema 3.1 — Temperatura média**

Peça três temperaturas ao usuário e calcule a média das medições.

A média deve ser calculada por:

$
\frac{T_1 + T_2 + T_3}{3}
$

Exiba uma mensagem e, depois, o resultado.

**🎯 Objetivo:** Calcular a média aritmética de três valores.


````python
# Gabarito do problema 3.1:

temperatura_1 = float(input("Digite a primeira temperatura: "))
temperatura_2 = float(input("Digite a segunda temperatura: "))
temperatura_3 = float(input("Digite a terceira temperatura: "))

media = (temperatura_1 + temperatura_2 + temperatura_3) / 3

print("Temperatura média:")
print(media)
````

### ✅ **Problema 3.2 — Área e perímetro de um retângulo**

Peça a base e a altura de um retângulo. Calcule:

$
\text{Área} = \text{Base} \times \text{Altura}
$

$
\text{Perímetro} = 2 \times (\text{Base} + \text{Altura})
$

Exiba os dois resultados.

**🎯 Objetivo:** Traduzir fórmulas matemáticas para expressões em Python.


````python
# Gabarito do problema 3.2:

base = float(input("Digite a base do retângulo: "))
altura = float(input("Digite a altura do retângulo: "))

area = base * altura
perimetro = 2 * (base + altura)

print("Área:")
print(area)

print("Perímetro:")
print(perimetro)
````

### ✅ **Problema 3.3 — Índice de Massa Corporal**

Peça o peso em quilogramas e a altura em metros de uma pessoa. Calcule o IMC:

$
\text{IMC} = \frac{\text{Peso}}{\text{Altura}^2}
$

Neste exercício, apenas calcule e exiba o valor. Não é necessário classificar o resultado.

**🎯 Objetivo:** Utilizar divisão e exponenciação em uma expressão.


````python
# Gabarito do problema 3.3:

peso = float(input("Digite o peso em quilogramas: "))
altura = float(input("Digite a altura em metros: "))

imc = peso / (altura ** 2)

print("IMC:")
print(imc)
````

> ## **Problema 4 — Divisão inteira, módulo e expressões em geral** 🟡

Praticando `//`, `%`, exponenciação e ordem de precedência.


### ✅ **Problema 4.1 — Convertendo minutos**

Peça uma quantidade total de minutos e converta-a para horas e minutos.

Utilize:

- `//` para obter as horas completas;
- `%` para obter os minutos restantes.

**🎯 Objetivo:** Aplicar divisão inteira e módulo.<br>
**📥 Exemplo de entrada:** `135`<br>
**📤 Exemplo de saída:**

```text
Horas:
2
Minutos:
15
```


````python
# Gabarito do problema 4.1:

total_minutos = int(input("Digite a quantidade total de minutos: "))

horas = total_minutos // 60
minutos = total_minutos % 60

print("Horas:")
print(horas)

print("Minutos:")
print(minutos)
````

### ✅ **Problema 4.2 — Separando cédulas**

Peça um valor inteiro em reais e informe quantas cédulas de R$ 10 podem ser utilizadas e quanto resta.

Utilize os operadores `//` e `%`.

**🎯 Objetivo:** Decompor um valor usando divisão inteira e resto.<br>
**📥 Exemplo de entrada:** `47`<br>
**📤 Exemplo de saída:**

```text
Cédulas de 10 reais:
4
Valor restante:
7
```


````python
# Gabarito do problema 4.2:

valor = int(input("Digite um valor inteiro em reais: "))

cedulas_de_10 = valor // 10
valor_restante = valor % 10

print("Cédulas de 10 reais:")
print(cedulas_de_10)

print("Valor restante:")
print(valor_restante)
````

### ✅ **Problema 4.3 — Prevendo resultados**

Antes de executar, tente descobrir o resultado de cada expressão.

**🎯 Objetivo:** Compreender a ordem de precedência dos operadores.<br>
**📌 Instrução:** Escreva cada resultado à direita do caractere `#`.


````python
# Gabarito do problema 4.3:

# Os comentários mostram os resultados previstos.
print(5 + 3 * 2)       # 11
print((5 + 3) * 2)     # 16
print(2 ** 3 + 1)      # 9
print(10 / 2 + 3)      # 8.0
print(10 / (2 + 3))    # 2.0
print(17 // 5)         # 3
print(17 % 5)          # 2
````

### ✅ **Problema 4.5 — Calculando...**

Construa um programa que peça:

- o nome de uma pessoa;
- três temperaturas medidas ao longo do dia;
- a quantidade de horas entre a primeira e a última medição.

O programa deve calcular:

- a temperatura média;
- a diferença entre a maior e a menor hora de registro, representada pela quantidade informada;
- o número total de minutos entre as medições.

Ao final, exiba:

```text
Nome da pessoa:
[nome digitado]

Temperatura média:
[resultado]

Intervalo em minutos:
[resultado]
```

**🎯 Objetivo:** Integrar entrada de dados, variáveis, conversões de tipos e operações aritméticas.


````python
# Gabarito do problema 4.5:

nome = input("Digite o nome da pessoa: ")

temperatura_1 = float(input("Digite a primeira temperatura: "))
temperatura_2 = float(input("Digite a segunda temperatura: "))
temperatura_3 = float(input("Digite a terceira temperatura: "))

intervalo_horas = float(
    input("Digite a quantidade de horas entre a primeira e a última medição: ")
)

temperatura_media = (
    temperatura_1 + temperatura_2 + temperatura_3
) / 3

intervalo_minutos = intervalo_horas * 60

print("Nome da pessoa:")
print(nome)

print("Temperatura média:")
print(temperatura_media)

print("Intervalo em minutos:")
print(intervalo_minutos)
````
