### 📝 **Problema 1.3**

**🎯 Objetivo:** Com base no exercício anterior, interpole os dados para fazer o tratamento dos dados e verifique novamente.

```python
# TODO: trate os dados faltantes
# Dica: use o método .interpolate()
df = df.interpolate(method='linear')

# TODO: Visualize informações da tabela
df.info()
```

### 📝 **Problema 1.4**

**🎯 Objetivo:** Crie uma visualização bidimenional utilizando as colunas x e y para visualizar o trajeto do `Albert`.

**Nota:** Utiliza a biblioteca `matplotlib.pyplot` e seus métodos para criar a visualização.

```python
# TODO: faça uma visualização simples da trajetória do albert ao longo do tempo
# Dica: use as funções plt.plot() e plt.show()

plt.plot(df['x'], df['y'], color='blue', marker='o')

plt.title('Caminho do Albert')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')

plt.grid()

plt.show()
```

### 📝 **Problema 1.5**

**🎯 Objetivo:** Crie uma coluna na estrutura de dados que mostra a distância percorrida entre as posições.

```python
# Dica: crie colunas auxiliares df['dx'] e df['dy'] que representam a diferença entre cada posição de um eixo utilizando o método .diff()
# Exemplo:
# df['dx'] = df['x'].diff()
# Faça o mesmo para o eixo y
df['dx'] = df['x'].diff()
df['dy'] = df['y'].diff()

# TODO: crie a coluna que motra a distância percorrida entre cada passo
# Dica: utilize a formula de pitagora e a função np.sqrt() para auxiliar
df['distancia_passo'] = np.sqrt(df['dx']**2 + df['dy']**2)
```

### 📝 **Problema 1.6**

**🎯 Objetivo:** Crie uma visualização para mostrar a distância percorrida entre as posições ao longo do tempo.

```python
# TODO: faça uma visualização simples da trajetória do albert ao longo do tempo
# Dica: use as funções plt.plot() e plt.show()

plt.plot(df['tempo'], df['distancia_passo'], color='green', marker='o')

plt.title('Distância Percorrida por Passo ao Longo do Tempo')
plt.xlabel('Tempo')
plt.ylabel('Tamanho do Passo')

plt.grid()

plt.show()
```
