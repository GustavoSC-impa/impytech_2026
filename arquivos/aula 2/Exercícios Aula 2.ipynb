> ## **Problema 5 - Curiosidade e Desafio** 🟢

float e recursão

### 📝 **Problema 5.1**

O seguinte código deveria, idealmente, imprimir "True", correto? rode ele e veja o que acontece. Por que será que isso ocorre?


```python
decimal = 0.1 +0.2

print(decimal == 0.3) # idealmente, seria True
```

Isso ocorre pois não é possível representar infinitos números no computador, e aproximações são feitas. Logo, para comparar floats, prefira

`print(math.isclose(decimal, 0.3))`
