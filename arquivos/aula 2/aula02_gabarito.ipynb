### ✅ **Resolução — Problema 1.1**
Imagine que uma clínica está com dois estudos, onde cada um está precisando de voluntários que apresentam os seguintes quadros:
1 - Parkinson

2 - Alzheimer

Dito isso, você receberá de entrada o quadro de um paciente, e o seu programa deve responder:

"Estudo 1" caso ele tenha Parkinson
"Estudo 2" caso tenha Alzheimer
"Não participa do estudo" Caso contrário

**🎯 Objetivo:** Decidir a qual estudo um voluntário participará <br>

### ✅ **Resolução — Problema 2.1**
Suponha que ao longo do dia um medicamento tenha sido administrado diversas vezes para vários pacientes, em diferentes doses. Essas doses todas foram computadas em um computador. Ao final do dia, para sinalizar que a distribuição do remédio encerrou, o número zero , "0", é escrito na última linha.

**🎯 Objetivo:** Ler e percorrer as entradas do medicamento enquanto elas são diferentes de 0. Ao final, imprimir a soma total de medicamento distribuido naquele dia. <br>

**📥 Exemplo de Entrada:**
10
5
6
0

**📤 Exemplo de Saída:**
21

```python
"""
Usamos um loop while que continua rodando até que a quantidade seja 0.
A cada iteração, somamos o valor recebido e pedimos uma nova entrada.
"""
quantidade = int(input())

soma = 0

while (quantidade != 0):
  soma += quantidade
  quantidade = int(input())

print(soma)
```

> ## **Problema 3 - Laço + Condicional** 🟢
Unindo o que foi visto até agora

### ✅ **Resolução — Problema 3.1**
Um determinado hospital te envia o número de pacientes com suspeita de PAC ( Pneumonia Adquirida na Comunidade) que eles atenderam durante a semana. Em seguida, para cada paciente, é enviado quatro informações, cada uma em uma linha, contendo as seguintes informações:

1 - Se o paciente estava originalmente internado ( Com 'S' de "sim" e 'N' de 'não')
2 - Número inteiro representando há quantos dias o paciente sofre com os sintomas
3 - Número inteiro representando a frequência respiratória
4 - Float representando a saturação de Oxigênio

Seu trabalho é ler esses dados e imprimir quantos pacientes dessa suspeita tem PAC, dado que foi convencionado que o diagnóstico só é confirmado se:
1 - O paciente não estava originalmente internado ( ou seja, 'N')
2 - O paciente está sofrendo há no máximo 14 dias
3 - Frequência respiratória maior que 27
4 - A saturação de Oxigênio é menor ou igual a 0.9

**🎯 Objetivo:** Ler uma entrada do número de pacientes e suas informações médicas e retornar o número de pacientes com PAC naquele hospital. <br>

**📥 Exemplo de Entrada:**
6
N
20
18
0.95
...

**📤 Exemplo de Saída:**
Pacientes com PAC: 1

```python
"""
Para cada paciente na contagem total (usando for), lemos as 4 entradas e checamos todas as condições simultaneamente com o operador 'and'.
"""
pacientes = int(input("Quantidade de pacientes: "))

soma = 0

for i in range(pacientes):
  internado = input()
  dias = int(input())
  freq_resp = int(input())
  sat_ox = float(input())
  if (internado == "N") and (dias <=14) and (freq_resp > 27) and (sat_ox <=0.9):
    soma += 1
print("Pacientes com PAC:", soma)
```

### ✅ **Resolução — Problema 3.2**
Na UTI a frequência cardíada dos pacientes é medida através de um computador, que vai registrando em batimentos por minuto a FC, até o momento onde o aparelho é desligado ( ou seja, é registrado "0" no computador).

Dado que:
Bracardia : BPM < 60
Normal :  BPM >= 60 e BPM <=100
Taquicardia : BPM > 100

**🎯 Objetivo:** Escreva um programa que leia todos os registros feitos pelo computador e imprima, para cada leitura, se o paciente estava com bracardia, normal, taquicardia, ou se o aparelho foi desligado. <br>

**📥 Exemplo de Entrada:**
50
80
110
0

**📤 Exemplo de Saída:**
Bracardia
Normal
Taquicardia
Aparelho Desligado

```python
"""
Lemos a entrada continuamente em um loop while, verificando a condição de batimentos usando ifs e elifs. O loop se quebra ao receber '0'.
"""
bpm = int(input("Digite bpm: "))

while bpm != 0:
  if bpm < 60:
    print("Bracardia")
  elif bpm > 100:
    print("Taquicardia")
  else:
    print("Normal")
  bpm = int(input("Digite bpm: "))

print("Aparelho Desligado")
```

### ✅ **Resolução — Problema 4.1**
Dado a soma de peso em kg total de n pacientes, escreva e aplique uma função que retorne o peso médio com 2 casas decimais de precisão.

**🎯 Objetivo:** Escreva uma função que retorne o peso médio de n pacientes; <br>

**📥 Exemplo de Entrada:**
98
4

**📤 Exemplo de Saída:**
24.50

```python
"""
Criamos uma função def que recebe o peso e a quantidade, e retorna a divisão matemática.
Utilizamos a formatação "%.2f" para exibir apenas duas casas decimais no print final.
"""
def media (peso, quantidade):
  return peso / quantidade

peso = int(input("Escreva o peso total: "))
quantidade = int(input("Escreva a quantidade total: "))

resposta = media(peso, quantidade)

print("%.2f" % resposta)
```

> ## **Problema 5 - Curiosidade e Desafio** 🟢
float e recursão

### ✅ **Resolução — Problema 5.1**
O seguinte código deveria, idealmente, imprimir "True", correto? rode ele e veja o que acontece. Por que será que isso ocorre?

**🎯 Objetivo:** Analisar o comportamento de precisão dos valores tipo float em Python. <br>

**🧠 Explicação lógica:**<br>
Isso ocorre pois não é possível representar infinitos números no computador, e aproximações são feitas nas linguagens de programação. Logo, para comparar floats com precisão absoluta, prefira utilizar funções de tolerância embutidas (ex: `math.isclose()`).

```python
"""
Ao somar 0.1 e 0.2 em ponto flutuante, o resultado gerado possui imperfeições microscópicas (0.30000000000000004), retornando False na checagem bruta.
"""
decimal = 0.1 + 0.2

print(decimal == 0.3) # idealmente, seria True
```

### ✅ **Resolução — Problema 5.2 - Desafio de Funções Recursivas**
Imagine que existe um novo vírus que no dia 0 infectou apenas uma pessoa ( inicialmente). Cada pessoa doente infecta exatamente 3 pessoas novas (pertentences a geração "G") antes de se recuperar. Dito isso, na geração G, quantas pessoas infectadas tem ao total ? Considere que esse total será jutamente igual a (numero de pessoas recem infectadas) + (Pessoas infectadas no total na geração G-1)

**🎯 Objetivo:** Escreva uma função recursiva que retorne o número de pessoas infectadas ao total na geração G ( considerando que ninguém melhorou) <br>

**📥 Exemplo de Entrada:**
2

**📤 Exemplo de Saída:**
Ao total temos 13 infectados

```python
"""
Função recursiva: o caso base (geracao == 0) garante o fim da recursão e retorna 1.
Para outras gerações, o código calcula os novos casos (3 elevado à geração) e soma com a mesma função chamada para a geração anterior (geracao-1).
"""
def infectados (geracao):
  if geracao == 0:
    return 1
  else:
    novos = 3 ** geracao
    anterior = infectados(geracao - 1)
    return novos + anterior

g = int(input("Coloque aqui a geração: "))

resp = infectados(g)

print("Ao total temos", resp, "infectados")
```
