> ## **Conceito 1 - Operadores Lógicos**
Um **Operador Lógico** é uma operação que pode negar, avaliar ou combinar estruturas booleanas e como resultado, retorna um valor de verdadeiro ou falso

### **Passo 1.1 — Igual a `==`**
Sempre que quisermos comparar se duas variáveis, números, strings etc são iguais
nós usamos o operador lógico "==". É importante escrever duas vezes o operador pois,
caso contrário, não está sendo feita uma comparação de dois valores, e sim uma
atribuição.

Podemos atribuir, ou não, a uma variável o resultado lógico " verdadeiro" ou "falso"
proveniente da comparção com os operadores lógicos.


````python

a = 2 == 3# atribuindo à variável "a" o resultado booleano da comparação se "2 é igual a 3"

print(a) # deveria ser printado "False" pois é falso que 2 é igual a 3
print( 4 == 5) # Deveria ser printado "False"
print( 3== 3) # True

n1 = "oi"
n2 = "oi"
print ( n1 == n2) #"True"

p1 = 2
p2 = 2
print ( p1 == p2) # True

print ( "oii" == "oi") # "oii" é diferente de "oi". Deveria ser printado "False"

#Atenção! Cuidado com comparações que envolvem floating points. Curiosidades específicas:

decimal = 0.1 +0.2

print(decimal == 0.3) # idealmente, seria True
````

### **Passo 1.2 — Diferente de `!=`**
Lógica semelhante ao de " igualdade", mas a negação, ou seja, comparar para vermos se é diferente ( nos retornando True se é diferente, e False se é igual) é com o uso do " !=". Utilizaremos os mesmos exemplos de igualdade da célula anterior.
Observe como estão trocadas as respostas de "False" e "True".

````python

a = 2 != 3

print(a)
print( 4 != 5)
print( 3!= 3)

n1 = "oi"
n2 = "oi"
print ( n1 != n2)

p1 = 2
p2 = 2
print ( p1 != p2)

print ( "oii" != "oi")

#Atenção! Cuidado com comparações que envolvem floating points. Curiosidades específicas:

decimal = 0.1 +0.2

print(decimal != 0.3) # idealmente, seria True
````

### **Passo 1.7 — E `and`**

Também chamado de "conjunção", é o operador que "soma" ou "agrupa" estruturas booleanas. Para uma cláusula retornar "verdadeiro", ambas as estruturas booleanas unidas pelo `and` devem ser verdadeiras.

````python
print ( 5 != 5 and 5<4)
print ( 5== 5 and 5<4)
print( 5!= 5 and 5 > 4)
print ( 5 == 5 and 5 > 4)
````

### **Passo 1.8 — OU `or`**

Também chamado de "disjunção". Para uma cláusula retornar "verdadeiro", pelo menos uma das estruturas booleanas unidas pelo `or` deve ser verdadeira.

````python
print ( 5 != 5 or 5<4)
print ( 5== 5 or 5<4)
print( 5!= 5 or 5 > 4)
print ( 5 == 5 or 5 > 4)
````

> ## **Conceito 2 - Condicionais**

**Condicionais** são estruturas que mandam seu código realizar determinadas etapas caso uma condição foi satisfeita. Ele é estruturado com `if`,`elif` e `else`. Para isso, é importante que as condições do seu `if` sejam bem estruturadas e que o comando de ação esteja tabulado dentro do `if`, `elif` e `else`.

> **O que vamos aprender:**
> - Como e quando usar `if`, `elif` e `else`
> - Quando usar `if`s em cadeia e não `elif`
> - Condicionais dentro de condicionais

> **Exemplo Base da Subseção de Condicionais:**

Imagine que tenhamos o resultado do índice de Shannon da microbiota de um paciente e nossa função é classificar em "Alta Diversidade", "Diversidade Moderada" e "Baixa Diversidade" a partir do valor numérico retornado, onde convencionamos que:


Alta diversidade : >= 3

Diversidade Moderada : entre 2 e 3 ( sem incluir o 3)

Baixa diversidade : < 2


### **Passo 2.1 — Como e quando usar `if`, `elif`, `else`**

if : se a condição satisfazer

elif : caso não tenha sido satisfeitas as condições anteriores E satisfazer uma nova condição

else: caso não tenha satisfeito nenhums dos "if"s e "elif"s anteiores, faça isso

O "elif" é uma boa ferramenta pois o computador apenas testa a condição do "elif" se algum "if" anterior já não foi satisfeito.


````python
indice = float(input())

if indice >= 3.0:
  print('Alta Diversidade')
elif indice < 3.0 and indice >= 2.0:
  print ('Diversidade Moderada')
else:
  print('Baixa Diversidade')
````

### **Passo 2.2 — Quando usar `if`s em cadeia e não `elif`**

 Às vezes alguns problemas exigem que coloquemos dois "ifs" em sequência para uma eventual possibilidade das condições não serem " excludentes". Digamos por exemplo que pro nosso estudo queremos estudar apenas pessoas com indice entre 3 e 3.5, e entre 2.5 e 3. Podemos fazer o seguinte código para nos retornar se o paciente está apto para participar do estudo, assim como retornar qual nível de diversidade sua microbiota está:

````python
indice = float(input())

if indice <=3.5 and indice >= 2.5:
  print('Pode Participar do Estudo')
if indice >= 3.0:
  print('Alta Diversidade')
elif indice < 3.0 and indice >= 2.0:
  print ('Diversidade Moderada')
else:
  print('Baixa Diversidade')
````

### **Passo 2.3 — Condicionais dentro de condicionais**

Colocamos condicionais dentro de condicionais quando, dado que uma condição foi satisfeita, queremos testar outras " sub-satisfações" / realizações. Podemos usar o mesmo exemplo da seção anterior:

````python
indice = float(input())

if indice >= 3.0:
  print('Alta Diversidade')
  if indice <=3.5:
    print ('Pode Participar do Estudo')
  else:
    print('Nao Pode Participar do Estudo')
elif indice < 3.0 and indice >= 2.0:
  print ('Diversidade Moderada')
  if indice >=2.5:
    print ('Pode Participar do Estudo')
  else:
    print('Nao Pode Participar do Estudo')
else:
  print('Baixa Diversidade')
  print('Nao Pode Participar do Estudo')
````

> ## **Conceito 3 - Laços de Repetição**
**Laços de Repetição** são estruturas que repetem comandos de código enquanto uma condição é verdadeira ou por uma quantidade "n" de vezes. Esses laços permitem que o usuário escreva o bloco de comando apenas uma vez ( e não "n" vezes).

> **O que vamos aprender:**
> - Por que Laço de Repetição é Útil
> - `for` e seus parâmetros
> - `while`
> - Operador de parada `break`

> **Exemplo Base da Subseção de Condicionais:**

>Nos exemplos abaixo, de cada uma das subseções, será apresentada uma adaptação do cenário onde temos "n" voluntários para o estudo de um medicamento, e de acordo com o número do voluntário ( a senha dele, que será um inteiro de 1 a n), será decidido se ele receberá placebo ou o medicamento que está sendo testado.

### **Passo 3.1 — Por que Laço de Repetição é Útil**

> Para ilustrar a utilidade dos laços de repetição, montaremos dois códigos que cumprem o mesmo objetivo, porém um usando laço de repetição e outro não. O objetivo é escrever em ordem quais voluntários receberão o medicamento, e quais não receberão, dado que temos 10 voluntários, e que os voluntários com senhas pares devem receber um placebo enquanto aqueles com senhas ímpares devem receber o medicamento.

````python


print('Paciente 1 recebe: Medicamento')
print('Paciente 2 recebe: Placebo')
print('Paciente 3 recebe: Medicamento')
print('Paciente 4 recebe: Placebo')
print('Paciente 5 recebe: Medicamento')
print('Paciente 6 recebe: Placebo')
print('Paciente 7 recebe: Medicamento')
print('Paciente 8 recebe: Placebo')
print('Paciente 9 recebe: Medicamento')
print('Paciente 10 recebe: Placebo')
````

> **Comentários sobre o código acima:**
> Observe que parece pouco prático escrever manualmente para cada um dos 10 indivíduos. E se fossem 100 ? A chance de digitar e se confundir também aumentam.

````python
for i in range (1,11):
  if i%2 == 1:
    print('Paciente',i,'recebe: Medicamento')
  else:
    print('Paciente',i,'recebe: Placebo')
````

> **Comentários sobre o código acima:**
> Note que é bem mais fácil de generalizar esse código para números de voluntários maiores. Se quisessemos para 100 voluntários, bastaria colocar `range(1,101)` como veremos na próxima subseção.

### **Passo 3.2 — `for`e seus parâmetros**

> É um laço onde tem três parâmetros:
> (inicio, fim, passo).

> Esses três parametros ficam em  `for i in range(inicio,fim,passo)`de forma que "i" vai assumindo valores a partir de "inicio" até "fim - 1", e avança de acordo com o passo. Se apenas um valor for informado em  `for i in range(n)`, então "n" é o fim, e o início é, por padrão "0" , e o passo é 1 ( ou seja, o loop vai rodar as n vezes pois começa em 0 e vai de 1 em 1  até n-1). Se dois valores forem informados, ou seja, `for i in range(m,n)`então "m" é o início e "n" o fim. Quando os três valores são informados, é da forma (inicio,fim,passo).
> O que está dentro do laço ( ou seja, o que está tabulado) é repetido a quantidade de vezes que o "i" poderia percorrer esse intervalo do "range".
> Observação: "i" é o nome de uma variável. Poderíamos nomear "i" de qualquer outra letra/nome. A seguir, será apresentado alguns exemplos, mudando os diferentes parâmetros do `range`. Ao final, uma adaptação do problema em "Conteito 3" será apresentada para contextualizar o `for` com seus diferentes parâmetros.

````python
for i in range (10):
  print (i)
````

> **Comentários sobre o código acima:**

> "i" começa em 0 e vai até 10-1 = 9, como explicado anteriormente. Logo, se ao invés de `print(i)`, tivéssemos colocado uma outra estrutura de comandos, esses comandos teriam sido feitos 10 vezes.

````python
for i in range(3,10):
  print (i)
````

> **Comentários sobre o código acima:**

> "i" começa em 3 e vai até 9, como explicado anteriormente.

````python
for i in range( 3, 15,2):
  print (i)
````

> **Comentários sobre o código acima:**

> "i" começa em 3 e vai até 14, porém com passo 2, ou seja, pula de 2 em 2. Como o último número possível do intervalo é 14, e 14 não é alcancável pulando de 2 em 2 a paritr de 3, então o último número printado é o 13.  

####**Exemplo Contextualizado**


Para exemplificar os outros parâmetros do `for`, digamos que um estudo clínico disponha de 100 voluntários. No entanto, o remédio a ser testado é muito caro, logo só poderemos testar em 10 pacientes. Por tanto, foi determinado que seriam escolhidos o candidato 25 até o candidato 43, de forma que, se o número do candidato for ímpar, ele deverá receber o remédio ( que está sendo testado). Nenhum dos demais candidatos receberão medicamento. Logo, a estrutura de loop  que determinará qual paciente tomará o medicamento será definido pelo loop:

````python
for i in range(25,45,2): # Ao inves de testar a paridade, podemos apenas pular de dois em dois ( ímpar em ímpar)
  print ('Paciente',i,'deve tomar o medicamento')
````

### **Passo 3.3 — `while`**

O `while`é um loop que roda enquanto uma condição for verdadeira. Portanto, é muito importante que a cada interação ( cada vez que o loop é repetido) haja alguma atualização no item que está sendo comparado. Caso não haja essa atualização, o loop se torna infinito.

> A seguir, mostraremos dois exemplos de como atualizar o comparador de um loop, seguindo a mesma contextualização anterior.

````python
contador_de_medicamentos = 0;
paciente = 25

while contador_de_medicamentos < 10:
  if paciente%2 == 1:
    contador_de_medicamentos += 1
    print('Paciente',paciente,'deve tomar o medicamento')
  paciente+=1
````

> **Explicação do código acima:**

> *  em `contador_de_medicamentos += 1` estamos fazendo a atualização necessária para garantir que não entrará em um loop infinito, uma vez que só sai desse loop quando `contador_de_medicamentos < 10`se torna falso, ou seja, quando `contador_de_medicamentos` passa a ser maior ou igual que 10.
> * Apenas atualizamos caso o número do paciente seja ímpar para seguir as mesmas instruções do exemplo anterior. Mas aumentamos o número do paciente sempre, tal qual como no `for`.  

### **Passo 3.4 — Operador de parada `break`**

> Um bom código, em geral, consegue ser programado para parar sem o comando do `break`. No entanto, às vezes pode parecer tentador colocar uma condição de parada no loop. Para isso, usamos o comando `break`.

> Uma observação importante é que o `break`apenas interrompe o loop mais imediato, ou seja, caso tivéssemos que o loop 2 está dentro do loop 1, e o `break` foi acionado dentro do loop 2 ( o laço mais interno) então o laço 2 será quebrado, mas o laço externo continuará "rodando".

Exemplos de uso de `break`:

````python
contador = 0

while contador <= 9:
  print(contador)
  if contador == 7:
    break
  contador+=1
````

### **Passo 4.1 — Como criar uma função**

> Para criar uma função, defina sempre ela antes do seu código principal.

> Uma função é declarada da seguinte forma:


```
def nome_função ( parametros):
    operações
    operações
    return(resultado)
```

A seguir, veremos:



> 1.   Criação de Função Com Parâmetro.
2.   Criação de Função Sem Parâmetro.
3.   Criação de Função Sem Retorno.



#### **Passo 4.1.1 — Criação de Função Com Parâmetro**

O exemplo será de uma função que recebe dois valores ( dois parâmetros) e retorna a média entre eles.

````python
def media(numero_1, numero_2):
  return (numero_1+numero_2)/2
````

> **Comentários sobre o código acima:**

> Em python não é necessário escrevermos a natureza da variável dos parâmetros, nem do retorno. Porém é uma boa prática de programação (para futuros leitores do código, ou para debugar o código mais facilmente) é informar o tipo de variável a ser usada na entrada e o tipo de variável retornada. Esse informe pode ser feito através de docstrings ou com anotações de tipo. Aqui vai um exemplo  com anotações de tipo e em seguida com docstring :


````python
def media(numero_1: float, numero_2 : float) -> float:
  return (numero_1+numero_2)/2
````

````python
def media(numero_1, numero_2):
  """
    Calcula a média entre dois números.

    Fórmula dada por: (número 1 + número 2)/2

	Parâmetros:
    numero_1 (float): número 1.
    numero_2 (float): número 2.

    Retorna:
    float : A média dos dois números.
    """
  return (numero_1+numero_2)/2
````

> **Comentários sobre os códigos acima:**

> Observe que a docstring permite dar um pouco mais de informação como o objetivo da função; Por isso, no resto dessa seção, usaremos a docstring.

#### **Passo 4.1.2 — Criação de Função Sem Parâmetro**

### **Passo 4.3 — Exemplo de função dentro de função**

> Se uma função "a" for usar a função "b", então a função "b" deve ser escrita e definida antes da função "a".


````python
def soma(a,b):

  """
    Calcula a soma entre dois números.

    Parâmetros:
    a (float): número 1.
    b (float): número 2.

    Retorna:
    float : A soma dos dois números.
    """
  return a+b

def media(a,b):
  """
    Calcula a média entre dois números.

	  Parâmetros:
    a (float): número 1.
    b (float): número 2.

    Retorna:
    float : A média dos dois números.
    """
  s = soma(a,b)
  return s/2

n1 = int(input())
n2 = int(input())

resp = media(n1,n2)
print(resp)
````
