> ## **Problema 1 - Acesso e análise de listas** 🟢
Praticando índices, fatiamento, verificação de elementos e funções aplicadas a listas.

### ✅ **Resolução — Problema 1.1**
Considere a lista de temperaturas fornecida na célula de código. Crie e exiba listas de fatiamento específicos.

🎯 **Objetivo:** Praticar diferentes formas de fatiamento de listas.<br>

🧠 **Explicação lógica:**
Utilizamos o fatiamento `[início:fim]` para extrair partes específicas da lista e `[:]` para criar uma cópia exata.

```python
"""
Utilizamos o fatiamento [início:fim] para extrair partes específicas da lista e [:] para criar uma cópia exata.
"""
temperaturas = [22.5, 23.1, 24.0, 25.3, 24.8, 23.9, 22.7]

print(temperaturas[:3])
print(temperaturas[-3:])
print(temperaturas[2:6])

copia_temperaturas = temperaturas[:]
print(copia_temperaturas)
```

### ✅ **Resolução — Problema 1.2**
Considere a lista fornecida. Use fatiamento para criar e exibir sublistas com passos (steps) diferentes.

🎯 **Objetivo:** Utilizar o terceiro parâmetro do fatiamento, chamado de passo.<br>

🧠 **Explicação lógica:**
Acrescentamos o terceiro parâmetro no fatiamento `[início:fim:passo]` para pular elementos ou inverter a lista com passos negativos.

```python
"""
Acrescentamos o terceiro parâmetro no fatiamento [início:fim:passo] para pular elementos ou inverter a lista com passos negativos.
"""
dados = [10, 20, 30, 40, 50, 60, 70, 80, 90]

print(dados[::2])
print(dados[::3])
print(dados[::-1])
```

### ✅ **Resolução — Problema 1.3**
Considere a lista de quantidades de Unidades Formadoras de Colônias (UFC). Exiba dados estatísticos básicos dessa lista.

🎯 **Objetivo:** Aplicar `len()`, `max()`, `min()` e `sum()` em uma lista numérica.<br>

🧠 **Explicação lógica:**
Usamos funções embutidas (built-ins) do Python para calcular tamanho, valores extremos, soma e, consequentemente, a média da lista.

```python
"""
Usamos funções embutidas (built-ins) do Python para calcular tamanho, valores extremos, soma e, consequentemente, a média da lista.
"""
amostras_ufc = [50000, 72000, 68000, 91000, 45000, 83000]

print(len(amostras_ufc))
print(max(amostras_ufc))
print(min(amostras_ufc))
print(sum(amostras_ufc))
print(sum(amostras_ufc) / len(amostras_ufc))
```

### ✅ **Resolução — Problema 1.4**
Considere a lista de códigos fornecida e verifique informações sobre elementos nela.

🎯 **Objetivo:** Verificar a presença ou a ausência de elementos em uma lista.<br>

🧠 **Explicação lógica:**
Os operadores `in` e `not in` retornam valores booleanos (True/False) indicando se um elemento está ou não presente na lista.

```python
"""
Os operadores in e not in retornam valores booleanos (True/False) indicando se um elemento está ou não presente na lista.
"""
codigos = [7, 15, 21, 42, 58, 73]

print(42 in codigos)
print(100 in codigos)
print(15 not in codigos)
```

🧠 **Explicação lógica:**

Apesar da imutabilidade, as tuplas funcionam com a mesma lógica de índices `[posição]` e fatiamentos `[início:fim]` das listas.

> ## **Problema 4 - Dicionários** 🟡
Consultando chaves, valores e pares, além de remover informações.

### ✅ **Resolução — Problema 4.1**
Considere o dicionário de um paciente e exiba seus componentes de forma desmembrada.

🎯 **Objetivo:** Consultar as diferentes partes de um dicionário.<br>

🧠 **Explicação lógica:**
Os métodos `.keys()`, `.values()` e `.items()` separam o dicionário em visualizações focadas em chaves, valores e pares tuplados, respectivamente.

```python
"""
Os métodos .keys(), .values() e .items() separam o dicionário em visualizações focadas em chaves, valores e pares tuplados, respectivamente.
"""
paciente = {
    'nome': 'Rafael',
    'idade': 28,
    'altura': 1.50,
    'peso': 99.67
}

print(paciente.keys())
print(paciente.values())
print(paciente.items())
```

### ✅ **Resolução — Problema 4.2**
Utilizando o dicionário fornecido, analise a veracidade das chaves e valores armazenados.

🎯 **Objetivo:** Utilizar `len()` e `in` corretamente em dicionários.<br>

🧠 **Explicação lógica:**
Por padrão, o operador `in` busca nas chaves de um dicionário. Para buscar nos valores, devemos forçar a checagem em `.values()`.

```python
"""
Por padrão, o operador in busca nas chaves de um dicionário. Para buscar nos valores, devemos forçar a checagem em .values().
"""
paciente = {
    'nome': 'Rafael',
    'idade': 28,
    'altura': 1.75,
    'peso': 73.4
}

print(len(paciente))
print('idade' in paciente)
print('telefone' in paciente)
print('Rafael' in paciente.values())
```

O dicionário contém uma informação temporária chamada `"observacao"`. Remova-a dinamicamente.

🎯 **Objetivo:** Remover um par de chave e valor usando `.pop()`.<br>

🧠 **Explicação lógica:**
O método `.pop(chave)` no contexto de dicionários deleta o par referente à chave informada e retorna seu respectivo valor.

```python
"""
O método .pop(chave) no contexto de dicionários deleta o par referente à chave informada e retorna seu respectivo valor.
"""
exame = {
    'codigo': 'EX-204',
    'paciente': 'Luiza',
    'resultado': 8.7,
    'observacao': 'revisar cadastro'
}

removido = exame.pop('observacao')
print(removido)
print(exame)
```

### ✅ **Resolução — Problema 4.4**
Considere o dicionário fornecido e manipule sua visualização usando loop e concatenação de strings.

🎯 **Objetivo:** Percorrer os pares de um dicionário com `.items()`.<br>

🧠 **Explicação lógica:**
Iterando sobre `.items()`, conseguimos desempacotar diretamente `chave` e `valor` a cada ciclo do loop `for`.

```python
"""
Iterando sobre .items(), conseguimos desempacotar diretamente chave e valor a cada ciclo do loop for.
"""
produto = {
    'nome': 'Álcool em gel',
    'quantidade': 18,
    'preco': 12.5
}

for chave, valor in produto.items():
    print(f"{chave}: {valor}")

print(len(produto))
```

> ## **Problema 5 - Desafios integradores** 🔴
Combinando listas, tuplas e dicionários para organizar e analisar dados.

### ✅ **Resolução — Problema 5.1**
Considere a lista de resultados fornecida e crie um dicionário condensando suas propriedades estatísticas principais.

🎯 **Objetivo:** Organizar resultados em uma lista e resumir as informações em um dicionário.<br>

🧠 **Explicação lógica:**
Após tratar a lista usando os métodos de manipulação, extraímos as métricas calculadas guardando-as em chaves categorizadas no dicionário.

```python
"""
Após tratar a lista usando os métodos de manipulação, extraímos as métricas calculadas guardando-as em chaves categorizadas no dicionário.
"""
resultados = [73, 88, -1, 79, 84]

resultados.append(82)
resultados.extend([76, 91])
resultados.remove(-1)
resultados.sort()

resumo = {
    'quantidade': len(resultados),
    'menor': min(resultados),
    'maior': max(resultados),
    'soma': sum(resultados),
    'media': sum(resultados) / len(resultados)
}

for chave, valor in resumo.items():
    print(f"{chave}: {valor}")
```

### ✅ **Resolução — Problema 5.2**
Considere o dicionário cujas chaves são códigos e cujos valores são tuplas contendo nome e idade.

🎯 **Objetivo:** Integrar dicionários e tuplas em uma estrutura de cadastro.<br>

🧠 **Explicação lógica:**
As estruturas podem ser aninhadas livremente no Python. Aqui, recuperamos uma tupla atrelada a uma chave, e depois fatiamos/indexamos a tupla.

```python
"""
As estruturas podem ser aninhadas livremente no Python. Aqui, recuperamos uma tupla atrelada a uma chave, e depois fatiamos/indexamos a tupla.
"""
participantes = {
    'P01': ('Ana', 20),
    'P02': ('Bruno', 22),
    'P03': ('Carla', 19),
    'P04': ('Diego', 21)
}

print(participantes.keys())
print('P03' in participantes)

removido = participantes.pop('P02')
print(f"Nome: {removido[0]}, Idade: {removido[1]}")

print(participantes)
print(len(participantes))
```

> ## **Problema 6 - Manipulação de Arquivos** 🟡
Leitura, escrita e atualização de dados externos.

### ✅ **Resolução — Problema 6.1**
Você recebeu um arquivo chamado `triagem_pacientes.txt` contendo o nome dos pacientes que chegaram ao pronto-socorro. Leia esse arquivo e armazene os nomes em uma lista.

🎯 **Objetivo:** Praticar a leitura de arquivos e o armazenamento de dados em listas.<br>
📥 **Exemplo de Entrada:** `triagem_pacientes.txt`<br>
📤 **Exemplo de Saída:** `Total de pacientes em espera: 4`

🧠 **Explicação lógica:**
Ao abrir o arquivo com modo de leitura ('r'), o método `.readlines()` traz o conteúdo estruturado automaticamente como uma lista de linhas.

```python
"""
Ao abrir o arquivo com modo de leitura ('r'), o método .readlines() traz o conteúdo estruturado automaticamente como uma lista de linhas.
"""
with open('triagem_pacientes.txt', 'r', encoding='utf-8') as arquivo:
    linhas = arquivo.readlines()

print(f"Total de pacientes em espera: {len(linhas)}")
```

### ✅ **Resolução — Problema 6.2**
Crie um programa que pegue uma lista de dicionários com resultados de exames e escreva esses resultados em um arquivo chamado `resultados_exames.txt`.

🎯 **Objetivo:** Aprender a manipular dados e escrevê-los em arquivo no modo `w`.<br>
📥 **Exemplo de Entrada:** `[{'nome': 'Ana', 'glicemia': 90}, {'nome': 'Bruno', 'glicemia': 110}]`<br>
📤 **Exemplo de Saída:** O Arquivo `resultados_exames.txt` criado adequadamente.

🧠 **Explicação lógica:**
O modo de escrita ('w') cria o arquivo ou sobrescreve seu conteúdo. Iteramos sobre cada dicionário da lista registrando os resultados com quebras de linha (`\n`).

```python
"""
O modo de escrita ('w') cria o arquivo ou sobrescreve seu conteúdo. Iteramos sobre cada dicionário da lista registrando os resultados com quebras de linha (\n).
"""
lista_exames = [{'nome': 'Ana', 'glicemia': 90}, {'nome': 'Bruno', 'glicemia': 110}]

with open('resultados_exames.txt', 'w', encoding='utf-8') as arquivo:
    for exame in lista_exames:
        arquivo.write(f"Nome: {exame['nome']} | Glicemia: {exame['glicemia']} mg/dL\n")
```

### ✅ **Resolução — Problema 6.3**
Um novo paciente chegou tardiamente e seu resultado de exame precisa ser incluído no arquivo gerado no passo anterior.

🎯 **Objetivo:** Utilizar o modo `a` (append) para atualizar arquivos de registro.<br>
📥 **Exemplo de Entrada:** `Carlos`, `95`<br>
📤 **Exemplo de Saída:** O arquivo `resultados_exames.txt` atualizado com três linhas.

🧠 **Explicação lógica:**
Com o modo de adição ('a' - append), preservamos o conteúdo que já existia no `.txt` e adicionamos os novos dados inseridos de forma contínua.

```python
"""
Com o modo de adição ('a' - append), preservamos o conteúdo que já existia no .txt e adicionamos os novos dados inseridos de forma contínua.
"""
novo_paciente = "Carlos"
nova_glicemia = 95

with open('resultados_exames.txt', 'a', encoding='utf-8') as arquivo:
    arquivo.write(f"Nome: {novo_paciente} | Glicemia: {nova_glicemia} mg/dL\n")
```
