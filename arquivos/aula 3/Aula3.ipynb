### **1.1 — Listas**

Listas são sequências de valores. Uma de suas vantagens é que elas são mutáveis ( ou seja, se originalmente você montou sua lista como [1,2,3] mas quer trocar o "2" por outro número, como o "4", você consegue, e a lista se torna [1,4,3]). Outra vantagem das listas implementadas em python é que elas podem armazenar mais de um tipo de variável ao mesmo tempo, ou seja, podemos criar listas com inteiro, string e float, como  ["Maria", 76, 8.7]. #### **Passo 1.1.1 — Criação de Lista**

Pode ser declarada explicitamente ou pode ser construída à medida que entradas vão sendo lidas.

 "append" é o nome do comando para adicionar algo a sua lista. A sintaxe é:

noma_da_lista.append(elemento_ou_nome_da_variavel_que_quer_adicionar)

Essa adição de elemento ocorre ao final da lista;

```python
lista1 = [3,3,6,8,9,20]

lista2 = [3,6,'s','casa',8.9]

lista_vazia = [] #iniciando a lista vazia

for i in range(5):
  elemento = input() #lendo em formato de string
  lista_vazia.append(elemento) # adicionando ao final da lista

print("a lista que originalmente era vazia agora eh:", lista_vazia)

print ("lista1 =",lista1)
print("lista2 =", lista2)
```


#### **Passo 1.1.2 — Acesso de posição de lista**



Digamos que eu precise acompanhar o índice de massa de alguns pacientes, e para esse estudo, estamos considerando que as únicas informações que estão sendo estudadas no momento é sexo, idade e o índice de massa corporal. Dessa forma, foi estabelecia uma convenção onde todo paciente teria uma lista (ficha) onde a primeira informação é o seu gênero, a segunda a sua idade e a terceira, armazenaríamos o índice. Dessa forma, como poderíamos acessar, após armazenar essas informações, o índice de massa e corporal de um paciente? Se precisassemos apenas dessa informação ?


Podemos acessar através do índice da lista! Como sabemos que essa informação fica armazenada na terceira posição da lista, se quisermos acessar a terceira posição podemos fazer:

`a = lista[2]` ou `print(lista[2])` , se quisermos imprimir.

O primeiro índice de uma lista é o 0, depois o 1, 2, 3 e assim por diante.

Ou seja:

acessar primeira posição : lista[0]

acessar segunda posição : lista[1]

acessar terceira posição : lista[2]

acessar quarta posição : lista[3]

e pode acessar até a última posição, com:

acessar última posição: lista[tamanho_da_lista - 1]

**Atenção:** tentar acessar uma posição fora da lista resulta em erro! ( IndexError).



```python
ficha1 = ["Homem", 37, 0.2]

indice_massa = ficha1[2]

print(indice_massa)


print(ficha1[2])


print(ficha1[5]) # exemplo de erro por acesso de índice fora da lista

```

Observação: em python é possível usar índices negativos para percorrer a lista de trás para frente. Ou seja, o índice -1 se refere ao último elemento, -2 ao penúltimo... Mas a restrição sobre índice fora da lista ainda existe.

```python
print(ficha1[-1])
print (ficha1[-2])

print(ficha1[-6])
```


#### **Passo 1.1.2 — Trocando de Elemento ( Mutabilidade)**

Digamos agora que após um mês seja necessário atualizar o índice de massa e gordura do paciente de "ficha1" para 0.25. Uma vez que listas são mutáveis, é possível fazer a seguinte troca:

```python
'''
Fatiamento
'''

lista_original = ['a','b','c','d','e','f','g']

fatiamento1 = lista_original[1:4] # lista que contém os elementos a partir do indice "1" ate indice 3;

print("fatiamento 1 =",fatiamento1)
print("lista original se matém como:",lista_original)

fatiamento2 = lista_original[3:] #lista que começa no indice 3 ( ou seja, quarto elemento) e vai até o final

print("fatiamento 2 =",fatiamento2)

fatiamento3 = lista_original[:3]

print("fatiamento 3 =",fatiamento3)

fatiamento4 = lista_original[:]

print("fatiamento 4 ( copia de lista original) =",fatiamento4)

#agora caso quisessemos mudar a lista original com o fatiamento:

lista_original[1:3] = ['j','k']

print(lista_original)

# o que aconteceria se colocassemos mais elementos do que o permitido pelo intervalo 1:3? e se colocassemos menos ?

lista_original[1:3] = ['r','s','t','u']
print(lista_original)

lista_original[1:4] = ['p']
print(lista_original)



```

###Percorrer uma lista


Nos exemplos vistos até agora, ou imprimimos a lista inteira, ou acessamos apenas um elemento dessa lista ( atraveś do índice). Mas e se precisássemos percorrer todos os elementos individualmente de uma lista ? Imagine que um contador de colônias retorna a quantidade de Unidade Formadora de Colônias por volume de várias placas de petri, e que esses valores são armazenados em uma lista. Com isso, é necessário percorrer cada um desses valores para verificar quantos resultados estão dentro do esperado.³

Percorrer através dos índices viabiliza o usuário a usar as mesmas condições do "for" conforme descritas na aula 2 ( ou seja, poderia escolher qual espaçamento percorre os índices da lista e tem liberdade de decidir o intervalo dos índices que serão percorridos). No entanto, também é possível usar o operador "in" para, ao invés de percorrer por índice, percorrer por elemento da lista.

```python
data_UFC = [100000,50000,70000, 300000, 40000]
resultados = 0
for i in range(len(data_UFC)):
  elemento = data_UFC[i]
  #print(data_UFC[i])
  if elemento >= 50000 and elemento <= 100000:
    resultados +=1
print("resultados dentro do esperado =",resultados)
```

```python
for elemento in data_UFC:
  print(elemento)
```
