> ## **Conceito 1 - Variáveis Booleanas**
Uma **variável booleana** é um tipo de variável `bool` que pode ter dois valores: `True` ou `False`

> **O que vamos aprender:**
> - O que são variáveis booleanas
> - Operadores lógicos
> - Comparadores lógicos
> - Condicionais `if`, `elif` e `else`



### **Passo 1.1 — Variáveis booleanas**
Podemos criar uma variável booleana utilizando o tipo `bool`. Essa variável só pode ter **um de dois valores possíveis:** `True` ou `False` (**Atenção: sempre com a primeira letra maiúscula**)


````python
# =============================================
# PASSO 1.1 — Variáveis booleanas
# =============================================

P = True  # Tente também usar bool(1)
Q = False # Tente também usar bool(0)
P2 = bool(1)
Q2 = bool(0)
print(P)
print(Q)
print(P2)
print(Q2)
````

> 📝 Explicação do código acima:
> - `P = True` e `P2 = bool(1)` → Define as variáveis com o valor booleano `True`
> - `Q = False` e `Q2 = bool(0)` → Define as variáveis com o valor booleano `False`

### **Passo 1.2 — Operadores lógicos**
Um **Operador Lógico** é uma operação que pode negar, avaliar ou combinar estruturas booleanas e, como resultado, retorna um valor de verdadeiro ou falso. Primeiramente, temos três principais operadores lógicos, que correspondem aos **conectivos lógicos** vistos na aula:

| Conectivo | Python | Funcionamento | Exemplo |
| :---: | :---: | :---: | :---: |
| "OU" | `or` | Retorna `True` se ao menos um dos dois valores for `True` | `P or Q` |
| "E" | `and` | Retorna `True` apenas se os dois valores forem `True` | `P and Q` |
| "NÃO" | `not` | Sempre retorna o **valor oposto** | `not P` |

````python
# =============================================
# PASSO 1.2 — Operadores lógicos
# =============================================

# [PARTE 1] - Definindo as variáveis
P = True  # Experimente trocar os valores
Q = False # Experimente trocar os valores

# [PARTE 2] - Utilizando os operadores

print(P and Q) # True and False
print(P or Q)  # True or False
print(not P)   # not True
print(not Q)   # not False
````

> 📝 Explicação do código acima:
> - `P and Q` → Como um dos valores é `True` (nesse caso, `P`), então o conectivo `and` retornará `True`. Retornaria `False` apenas se `P` e `Q` fossem `False`
> - `P or Q` → Como um dos valores é `False` (nesse caso, `Q`), então o conectivo `and` retornará `False`. Retornaria `True` apenas se `P` e `Q` fossem `True` ao mesmo tempo
> - `not P` e `not Q` → Troca o valor lógico da expressão.

### **Passo 1.3 — Comparadores lógicos**
Os **comparadores lógicos** são um tipo de operador que podem ser utilizados para trabalhar com outros tipos de variáveis (como números, por exemplo). Os principais comparadores são:

| Comparador | Python | Exemplo |
| :---: | :---: | :---: |
| Igualdade | `==` | `1 == 2` |
| Diferente | `!=` | `1 != 2` |
| Maior | `>` | `2 > 1` |
| Maior ou igual | `>=` | `2 >= 1` |
| Menor | `<` | `1 < 3` |
| Menor ou igual | `<=` | `1 <= 3` |


````python
# =============================================
# PASSO 1.3 — Comparadores lógicos
# =============================================

print(1 == 2) # False
print(1 != 2) # True
print(2 > 1)  # True
print(2 >= 1) # True
print(1 < 3)  # True
print(1 <= 3) # True

# AGORA É SUA VEZ: Tente utilizar os comparadores lógicos para comparar duas variáveis, e veja o resultado
````

> 📝 Explicação do código acima:
> - `1 == 2` → O operador `==` verifica a igualdade. Como 1 não é igual a 2, o comparador retorna `False`.
> - `1 != 2` → O operador `!=` verifica se os valores são diferentes. Como 1 é de fato diferente de 2, o comparador retorna `True`.
> - `2 > 1` → O operador `>` verifica se o primeiro valor é estritamente maior que o segundo. Como 2 é maior que 1, o comparador retorna `True`.
> - `2 >= 1` → O operador `>=` verifica se o primeiro valor é maior **ou** igual ao segundo. Como 2 é maior que 1, a condição é satisfeita e retorna `True`.
> - `1 < 3` → O operador `<` verifica se o primeiro valor é estritamente menor que o segundo. Como 1 é menor que 3, o comparador retorna `True`.
> - `1 <= 3` → O operador `<=` verifica se o primeiro valor é menor **ou** igual ao segundo. Como 1 é menor que 3, a condição é satisfeita e retorna `True`.

> ## **Conceito 2 - Condicionais**

**Condicionais** são estruturas que mandam seu código realizar determinadas etapas caso uma condição foi satisfeita. Ele é estruturado com `if`,`elif` e `else`. Para isso, é importante que as condições do seu `if` sejam bem estruturadas e que o comando de ação esteja tabulado dentro do `if`, `elif` e `else`.

> **O que vamos aprender:**
> - Condicional `if else`
> - Condicionais múltiplas com `elif`



### **Passo 2.1 — Condicionais `if else`**
Podemos criar uma condicional simples utilizando `if(C)` (em português, **se**), onde `C` é uma **condição** (alguma variável booleana, ou expressão utilizando operadores lógicos). Se `C = True`, então o código irá executar o trecho que está tabulado ao comando `if()`. Além do `if()`, podemos utilizar também `else()` (em português, **senão**), que executará outro código caso a condição não seja satisfeita (`C = False`):

````python
# =============================================
# PASSO 2.1 — Condicionais if else
# =============================================

T = float(input("Qual a temperatura de Pedro? "))  # Tente digitar uma temperatura maior que 37,
                                                  # e depois uma menor que 37, e veja o que aparece
                                                  # em ambos os casos

if (T > 37):
    # Esse trecho será executado apenas se a temperatura for MAIOR que 37
    situacao = "Acima do normal"

else:
    # Caso a temperatura não seja maior que 37 (T <= 37), executa esse trecho
    situacao = "Normal"

print(situacao)
````

### **Passo 2.2 — Condicionais múltiplas com `elif`**
Quando precisamos fazer uma análise em que temos mais de duas possibilidades (`True` ou `False`), podemos utilizar `elif()` para passarmos várias condições, e o nosso programa executará o primeiro trecho de código em que a condição seja `True`

````python
# =============================================
# PASSO 2.2 — Condicionais múltiplas com elif
# =============================================

# Experimente digitar três valores diferentes de forma que:
# - Um deles seja MAIOR que 37
# - Um deles seja MENOR que 35
# - Um deles esteja entre 35 e 37 (35 <= T <= 37)
T = float(input("Qual a temperatura de Pedro?"))

if (T > 37):
    # Se T > 37, executa esse trecho
    situacao = "Acima do normal"

elif(T < 35):
    # Se T < 35, executa esse trecho
    situacao = "Abaixo do normal"

else:
    # Se nenhuma das condições acima for satisfeita (T está entre 35 e 37), executa esse trecho
    situacao = "Normal"

print(situacao)
````

> 📝 Explicação do código acima:
> - `T = float(input(...))` → Solicita que o usuário digite a temperatura de Pedro, converte esse texto digitado para um número decimal (`float`) e armazena na variável `T`.
> - `if (T > 37):` → Avalia a primeira condição. Se a temperatura fornecida for estritamente maior que 37, a variável `situacao` recebe o texto `"Acima do normal"`.
> - `elif(T < 35):` → Caso a condição anterior do `if` seja falsa, o programa testa essa nova condição. Se a temperatura for menor que 35, a `situacao` recebe `"Abaixo do normal"`.
> - `else:` → Se nenhuma das condições anteriores for satisfeita (ou seja, se `T` for 35, 37 ou estiver entre eles), a `situacao` recebe obrigatoriamente `"Normal"`.
> - `print(situacao)` → Imprime na tela a string que foi atribuída à variável `situacao` após a avaliação das condições.

> ## **Conceito 3 - Laços de Repetição**
**Laços de Repetição** são estruturas que repetem comandos de código enquanto uma condição é verdadeira ou por uma quantidade "n" de vezes. Esses laços permitem que o usuário escreva o bloco de comando apenas uma vez ( e não "n" vezes).

> **O que vamos aprender:**
> - Laços de repetição `for`
> - Laços de repetição `while`
> - Operador de parada `break`

### **Passo 3.1 — Laços de repetição `for`**

Os laços de repetição `for` usam a seguinte estrutura: <br><br> `for c in range(n):` <br><br> onde `c` é chamada de **variável de controle** e `range(n)` é nosso **controlador de passo**. Nesse caso, **n** indica o número de vezes que queremos que um trecho de código se repita.

````python
# =============================================
# PASSO 3.1 — Laços de repetição for
# =============================================

n = 3  # Tente mudar o número de repetições

for c in range(n):  # Será executado n vezes
    print("Olá!")
````

````python
# =============================================
# PASSO 3.1 — Laços de repetição for
# =============================================

n = 5

for c in range(n):
    print(c)  # Exibe o valor da variável de controle
````

````python
# =============================================
# PASSO 3.1 — Laços de repetição for
# =============================================

n = 5

for c in range(n):
    print(c ** 2) # Exibe o quadrado do valor da variável de controle
````

````python
# =============================================
# PASSO 3.1 — Laços de repetição for
# =============================================

for c in range(0, 5, 1):  # De 0 a 4, pulando de 1 em 1
    print(c)
````

````python
# =============================================
# PASSO 3.1 — Laços de repetição for
# =============================================

for c in range(1, 5, 1):  # De 1 a 4, pulando de 1 em 1
    print(c)
````

````python
# =============================================
# PASSO 3.1 — Laços de repetição for
# =============================================

for c in range(0, 6, 1): # De 0 a 5, pulando de 1 em 1
    print(c)
````

````python
# =============================================
# PASSO 3.1 — Laços de repetição for
# =============================================

for c in range(0, 6, 2): # De 0 a 5, pulando de 2 em 2
    print(c)
````

### **Classificador (múltiplos pacientes)**

````python
# =============================================
# PASSO 3.1 — Classificador (múltiplos pacientes)
# =============================================

for c in range(5):   # Repete o programa cinco vezes

   T = float(input("Qual a temperatura do Paciente?"))

   if (T > 37):
       situacao = "Acima do normal"
   elif(T < 35):
       situacao = "Abaixo do normal"
   else:
       situacao = "Normal"

   print(situacao)
````

````python
# =============================================
# PASSO 3.1 — Laços de repetição for
# =============================================

for paciente in range(5):       # inicio=0 (padrão) / fim=5 / passo=1 (padrão)
    print(paciente)

for temperatura in range(0,5):  # inicio=0 / fim=5 / passo=1 (padrão)
    print(temperatura)
````

### **Passo 3.2 — Laços de repetição `while`**

Diferente dos laços `for`, que executam um trecho de código uma quantidade determinada de vezes, podemos usar o laço `while` para executar um trecho de código **enquanto** uma dada condição for verdadeira, ou seja, uma quantidade **indeterminada de vezes**.

````python
# =============================================
# PASSO 3.2 — Laços de repetição while
# =============================================

n = 0

while(n<3):
    n = n + 1
    print(n)

print("Fim")
````

### **Classificador (usando while)**

````python
# =============================================
# PASSO 3.2 — Classificador (usando while)
# =============================================

T = float(input("Qual a temperatura do Paciente?"))

# Enquanto você não digitar temperatura zero, o código continuará rodando indefinidamente
while T != 0:  # Digite 0 para sair do laço while

   T = float(input("Qual a temperatura do Paciente?"))

   if (T > 37):
       situacao = "Acima do normal"
   elif(T < 35):
       situacao = "Abaixo do normal"
   else:
       situacao = "Normal"

   print(situacao)

print("Fim do programa")
````

### **Passo 3.3 — Operador de parada `break`**

> Um bom código, em geral, consegue ser programado para parar sem o comando do `break`. No entanto, às vezes pode parecer tentador colocar uma condição de parada no loop. Para isso, usamos o comando `break`.

Exemplos de uso de `break`:

````python
# =============================================
# PASSO 3.3 — Operador de parada break
# =============================================

n = 0

while(n < 100):
    n = n + 1
    print(n)

    if(n == 3):
        break

print("Fim")
````

> ## **Conceito 4 - Funções**
**Funções** Em programação podem ser interpretadas de forma análoga às funções matemáticas já conhecidas: Em geral, enviamos um valor de entrada para elas, e elas nos retornam um valor de saída. Funções em programação podem ou não ter parâmetros (o "valor de entrada" ) e podem ou não ter retorno ( "valor de saída").  Elas são úteis pois, em casos onde são sequências de "passo a passo" muito longos, ou em cenários onde serão utilizadas muitas vezes ao longo do cógido, a compreensão do funcionamento do cógido se torna mais limpa se elas forem definidas apenas uma vez e forem apenas chamadas ao longo do código. Os exemplos da seção devem ilustrar melhor o conceito de função.

> **O que vamos aprender:**
> - Criando funções
> - Passando parâmetros
> - Retornando valores

### **Passo 4.1 — Criando uma função**

> Para criar uma função, defina sempre ela antes do seu código principal. Uma função é declarada da seguinte forma:

```
def nome_função (parametros):
    operações
    operações
    return(resultado)
```



````python
# =============================================
# PASSO 4.1 — Criando uma função
# =============================================

def cumprimentar():
    print("Olá mundo!")

cumprimentar()
cumprimentar()
cumprimentar()
````

````python
# =============================================
# PASSO 4.1 — Criando uma função
# =============================================

def cumprimentar():
    print("Olá mundo!")
def tracos():
    print("----------")

tracos()
cumprimentar()
tracos()
````

### **Passo 4.2 — Passando parâmetros**

Na maioria das vezes, precisaremos de funções que funcionem com algum conjunto de **par[link text](https://)âmetros** que serão enviados quando a função for chamada

````python
# =============================================
# PASSO 4.2 — Passando parâmetros
# =============================================

def somar(n1, n2): # Dois parâmetros: n1 e n2
    print(n1 + n2)

somar(1,1)  # n1= 1 e n2= 1
somar(10,9) # n1= 10 e n2= 9
somar(-5,5) # n1= -5 e n2= 5
````

````python
# =============================================
# PASSO 4.2 — Passando parâmetros
# =============================================

def somar(n1, n2): # Dois parâmetros: n1 e n2
    print(n1 + n2)

soma = somar(1,1) # O valor (2) irá aparecer pois a função de soma EXIBE a soma dos números
print(soma)       # Essa linha irá exibir “None”
print(soma + 1)   # Essa linha gerará um erro
````

### **Passo 4.3 — Retorno de valores**

Para que possamos "olhar para uma função como um valor", semelhante as funções matemáticas em que você "insere" um número e recebe outro número, precisamos programar para que nossa função **retorne** algo

Nos exemplos anteriores, as funções que definimos apenas usavam `print()` para exibir um valor. Por isso, no exemplo acima, a variável `soma` exibia `None` quando era printada.

Para isso, usamos `return` no final da função, e o que queremos que ela retorne

````python
# =============================================
# PASSO 4.3 — Retorno de valores
# =============================================

def somar(n1, n2): # Dois parâmetros: n1 e n2
    return (n1 + n2) # Retorna a soma de dois números, ao invés de só exibir

soma = somar(1,1) # Agora, a nossa variável será realmente tratada como um número
print(soma)       # Diferente do caso anterior, o valor (2) irá aparecer por causa dessa linha
print(soma + 1)   # Como "soma" agora é um número, podemos somá-lo com outro número
````

> ## **Conceito 5 - Bibliotecas**

### **Passo 5.1 - Como importar e usar uma biblioteca em geral**

Ao usar uma biblioteca pela primeira vez em uma IDE pessoal, ou então ao usar uma biblioteca "fora do padrão" é possível que ela não tenha sido instalada. Para isso, usa-se geralmente o comando `pip install nome_biblioteca` no terminal do IDE. Cada biblioteca terá na sua documentação uma instrução de como instalá-la pela primeira vez.

Com a biblioteca já instalada, é necessário, no início o código, fazer a seguinte declaração:

`import nome_biblioteca`

ou

`import nome_biblioteca as abreviacao`

ou

`from nome_biblioteca import funcao_que_sera_usada`

A segunda opção é a mais comum, porém se apenas utilizaremos uma função de uma determinada biblioteca, também é comum utilizar o terceiro método de importação. A seguir, faremos os exemplos de importação com as próprias bibliotecas mencionadas. Durante o corpo do código, se quisermos utilizar uma função dessas bibliotecas, em geral fazemos: `biblioteca.funcao(seus_parametros)`. Observe os exemplos abaixo.

````python
# =============================================
# PASSO 5.1 — Como importar e usar uma biblioteca em geral
# =============================================

import math
from random import randint

print(math.pi)       # Acessa o valor “pi” da biblioteca “math”, que retorna uma aproximação de pi
print(randint(1,10)) # Usa a função “randint()” para gerar um inteiro aleatório entre 1 e 10
````
