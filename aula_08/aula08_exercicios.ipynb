###1.3
Os pacientes com a ficha preenchida seguem para alguns exames na unidade, e alguns casos levam a uma atualização de um conjunto dos dados previamente fornecidos. Com isso, o objeto precisa de uma atualização nas suas propriedades, visto que os exames mais recentes, para este caso, são aqueles de maior relevância. Então cada `Paciente` precisa de uma visualização dos seus dados diferente da ficha original, ou seja, eles devem possuir uma espécie de prontuário atualizado.

```python
"""
Atribua ao objeto o método de atualização dos dados.
Crie um método de visualização do conjunto de dados desejado do objeto com, no mínimo, estes dados: nome, idade, temperatura, dor de cabeça e dôr muscular.
obs: veja o que acontece ao chamar o método prontuário no método exame.
"""

class Paciente():

  unidade = 'IMPYSUS'
  bairro = 'PYTHON'

  def __init__(
      self,
      nome, idade,
      temp: float = 36.4,
      dc: bool = False,
      dm: int = 0
      ):

    self.nome = nome
    self.idade = idade

    self.temperatura = temp
    self.dor_cabeca = dc
    self.dor_musc = dm

  def geografia(self):
    print(self.unidade,self.bairro)

  def exame(self, temp, dc, dm):
    self.temperatura = temp
    self.dor_cabeca = dc
    self.dor_musc = dm
    self.prontuario()

  def prontuario(self):
    print("")
    print("Paciente:", self.nome,",",self.idade)
    print("Temperatura:", self.temperatura)
    print("Dor de cabeça:", self.dor_cabeca)
    print("Dor muscular:", self.dor_musc)
    print("")


pac_1 = Paciente("Fulano", 71)
pac_2 = Paciente("Ciclano", 45, 39.4)
pac_3 = Paciente("Beltrano", 31, dc = True)
pac_4 = Paciente("Fulano Jr", 20, 37.0, dm = 3)
pac_5 = Paciente("Ciclano Jr", 14)

pac_1.prontuario()
pac_2.prontuario()
pac_3.prontuario()
pac_4.prontuario()
pac_5.prontuario()

pac_5.exame(37.1, False, 3)

#pac_5.prontuario()
```
