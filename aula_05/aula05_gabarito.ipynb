> ## **Problema 2 - Tratamento de Sinais Vitais** 🟢

Nesta prática, você aplicará funções fundamentais para tratar uma tabela contendo os sinais vitais diários de pacientes em uma clínica.

Para isso, utilize os dados no arquivo `sinais_vitais.csv`.

### ✅ **Resolução — Problema 2.1**

**🎯 Objetivo:** Leia o arquivo `.csv` utilizando a biblioteca Pandas e exiba as primeiras linhas.

🧠 **Explicação lógica:**
Utilizamos a função `pd.read_csv()` para carregar os dados na memória e `.head()` para visualizar rapidamente as primeiras linhas do DataFrame.

```python
df_sinais = pd.read_csv('sinais_vitais.csv')
print(df_sinais.head())
```

### ✅ **Resolução — Problema 2.2**

**🎯 Objetivo:** Identifique a quantidade de valores ausentes em cada coluna utilizando `.isnull().sum()`.

🧠 **Explicação lógica:**
O método `.isnull()` cria um "dataframe booleano" de valores nulos, e ao aplicarmos `.sum()`, contamos quantos valores verdadeiros (nulos) existem por coluna.

```python
print(df_sinais.isnull().sum())
```

### ✅ **Resolução — Problema 2.3**

**🎯 Objetivo:** Preencha o dado faltante na coluna `Idade` com um valor numérico fixo de sua escolha utilizando `.fillna()`.

🧠 **Explicação lógica:**
Utilizamos `.fillna()` especificando um valor (ex: 30) para preencher os buracos apenas na coluna de Idade.

```python
df_sinais['Idade'] = df_sinais['Idade'].fillna(30)
```

### ✅ **Resolução — Problema 2.4**

**🎯 Objetivo:** Remova a linha que contém o valor nulo na coluna `Frequencia_Cardiaca` utilizando a função `.dropna()`.

🧠 **Explicação lógica:**
Usando `.dropna(subset=[...])`, garantimos que apenas as linhas que possuem valores ausentes nessa coluna específica sejam removidas.

```python
df_sinais = df_sinais.dropna(subset=['Frequencia_Cardiaca'])
```

### ✅ **Resolução — Problema 2.5**

**🎯 Objetivo:** Preencha os valores ausentes na coluna `Temperatura` utilizando a média das temperaturas da tabela.

🧠 **Explicação lógica:**
Primeiro calculamos a média da coluna com `.mean()` e usamos esse valor como parâmetro no método `.fillna()`.

```python
media_temp = df_sinais['Temperatura'].mean()
df_sinais['Temperatura'] = df_sinais['Temperatura'].fillna(media_temp)
```

### ✅ **Resolução — Problema 2.6**

**🎯 Objetivo:** Renomeie a coluna `Frequencia_Cardiaca` para `BPM` utilizando um dicionário na função `.rename()`.

🧠 **Explicação lógica:**
O método `.rename()` aceita um dicionário no argumento `columns` onde a chave é o nome antigo e o valor é o nome novo.

```python
df_sinais = df_sinais.rename(columns={'Frequencia_Cardiaca': 'BPM'})
```

### ✅ **Resolução — Problema 2.7**

**🎯 Objetivo:** Crie um novo DataFrame filtrando apenas os pacientes que estão com febre (temperatura maior ou igual a 37.5).

🧠 **Explicação lógica:**
Aplicamos uma máscara condicional no DataFrame para retornar apenas os dados em que a condição seja verdadeira.

```python
df_febre = df_sinais[df_sinais['Temperatura'] >= 37.5]
print(df_febre.head())
```

### ✅ **Resolução — Problema 2.8**

**🎯 Objetivo:** Construa um **Gráfico de Barras** simples para relacionar a coluna `Paciente` (eixo X) com a sua `Temperatura` (eixo Y).

🧠 **Explicação lógica:**
Utilizamos `plt.bar()` passando os dados dos eixos X e Y correspondentes, gerando a visualização gráfica.

```python
plt.bar(df_sinais['Paciente'], df_sinais['Temperatura'])
plt.title('Temperatura por Paciente')   # Título do gráfico
plt.xlabel('Paciente')                  # Título do eixo X
plt.ylabel('Temperatura')               # Título do eixo Y
plt.show()                              # Exibe o gráfico
```

> ## **Problema 3 - Análise de Custos e Internações** 🟡

Nesta prática, você deverá preparar os dados de internações hospitalares para uma análise estatística básica, lidando com inconsistências e criando novas categorias.

Para este problema, use o arquivo `internacoes_hospitalares.csv`.

### ✅ **Resolução — Problema 3.1**

**🎯 Objetivo:** Carregue o arquivo num DataFrame e exiba um resumo estatístico numérico da tabela utilizando `.describe()`.

🧠 **Explicação lógica:**
Carregamos a base via `read_csv` e utilizamos `.describe()` para gerar métricas essenciais como média, desvio padrão, min e max das colunas numéricas.

```python
df_internacoes = pd.read_csv('internacoes_hospitalares.csv')
print(df_internacoes.describe())
```

### ✅ **Resolução — Problema 3.2**

**🎯 Objetivo:** Remova as linhas em que a coluna `Setor` possua dados nulos especificando o subconjunto no método `.dropna()`.

🧠 **Explicação lógica:**
Semelhante ao exercício de sinais vitais, utilizamos o parâmetro `subset` para focar a remoção de nulos em uma coluna específica.

```python
df_internacoes = df_internacoes.dropna(subset=['Setor'])
```

### ✅ **Resolução — Problema 3.3**

**🎯 Objetivo:** Preencha o valor vazio na coluna `Dias_Internado` utilizando o método de interpolação linear (`.interpolate()`).

🧠 **Explicação lógica:**
O método `.interpolate(method='linear')` tenta preencher o valor nulo acompanhando uma progressão linear entre o valor anterior e o posterior da tabela.

```python
df_internacoes['Dias_Internado'] = df_internacoes['Dias_Internado'].interpolate(method='linear')
```

### ✅ **Resolução — Problema 3.4**

**🎯 Objetivo:** Preencha os valores vazios de `Custo_Estimado` utilizando a mediana ou média.

🧠 **Explicação lógica:**
Pegamos a mediana da coluna via `.median()` para preencher os dados ausentes. A mediana geralmente é melhor do que a média quando a base pode sofrer variações bruscas (outliers).

```python
mediana_custo = df_internacoes['Custo_Estimado'].median()
df_internacoes['Custo_Estimado'] = df_internacoes['Custo_Estimado'].fillna(mediana_custo)
```

### ✅ **Resolução — Problema 3.5**

**🎯 Objetivo:** Crie uma função personalizada em Python que classifique o custo da internação como `"Alto Risco Financeiro"` se passar de `10.000`, ou `"Custo Normal"` caso contrário.

🧠 **Explicação lógica:**
Criamos uma função simples utilizando o condicional `if/else` que retornará a classificação correta dado um valor numérico.

```python
def classificar_custo(valor):
    if valor > 10000:
        return "Alto Risco Financeiro"
    else:
        return "Custo Normal"
```

### ✅ **Resolução — Problema 3.6**

**🎯 Objetivo:** Aplique essa função na coluna `Custo_Estimado` utilizando o método `.apply()` e armazene o resultado em uma nova coluna chamada `Status_Custo`.

🧠 **Explicação lógica:**
O método `.apply()` varre a coluna enviando cada elemento para a nossa função, gerando os resultados que vão formar a nossa nova coluna do dataframe.

```python
df_internacoes['Status_Custo'] = df_internacoes['Custo_Estimado'].apply(classificar_custo)
print(df_internacoes.head())
```

### ✅ **Resolução — Problema 3.7**

**🎯 Objetivo:** Agrupe os dados por `Setor` utilizando o método `.groupby()` para descobrir a média de `Dias_Internado` em cada departamento.

🧠 **Explicação lógica:**
Ao utilizar `.groupby('Setor')`, separamos os dados nos seus respectivos grupos, aplicamos a coluna que queremos observar e encerramos chamando uma métrica, no caso a média `.mean()`.

```python
media_dias = df_internacoes.groupby('Setor')['Dias_Internado'].mean()
print(media_dias)
```

### ✅ **Resolução — Problema 3.8**

**🎯 Objetivo:** Construa um **Gráfico de Dispersão**, colocando `Dias_Internado` no eixo X e o `Custo_Estimado` no eixo Y, para visualizar a correlação entre o tempo de estadia e o valor da internação.

🧠 **Explicação lógica:**
Utilizamos a função `plt.scatter()` que mapeia dois valores correlacionados através de pontos no plano.

```python
plt.scatter(df_internacoes['Dias_Internado'], df_internacoes['Custo_Estimado'], alpha=0.5)
plt.title('Dias Internado vs. Custo Estimado')  # Título do gráfico
plt.xlabel('Dias Internado')                    # Título do eixo X
plt.ylabel('Custo Estimado')                    # Título do eixo Y
plt.show()                                      # Exibe o gráfico
```

> ## **Problema 4 - Expressão Genética** 🟢

Suponha que você e seu grupo de pesquisa estão fazendo um levantamento para descobrir uma relação entre a expressividade genética e o tumor.
Neste levantamento, vocês escolheram dois genes específicos: `TP53` e `BRCA1`.

### ✅ **Resolução — Problema 4.1**

**🎯 Objetivo:** Usando a biblioteca pandas leia o arquivo `dados_expressao_genes.csv`, visualize de forma simples a tabela.

🧠 **Explicação lógica:**
O arquivo CSV é carregado na memória como um DataFrame utilizando a função `pd.read_csv()`. Em seguida, o método `.head()` é empregado para mostrar uma prévia organizada das primeiras linhas.

```python
"""
O arquivo CSV é carregado na memória como um DataFrame utilizando a função pd.read_csv().
Em seguida, usamos .head() para checar os dados.
"""
df = pd.read_csv('./dados_expressao_genes.csv')

print(df.head())
```

### ✅ **Resolução — Problema 4.2**

**🎯 Objetivo:** Usando uma filtragem simples, faça um histograma comparando a quantidade de amostras de cada grupo de controle para o genes `TP53`.

🧠 **Explicação lógica:**
Aplicamos uma máscara booleana para filtrar a tabela apenas onde a coluna "Genes" corresponde a "TP53". Após a filtragem, utilizamos `.value_counts()` para somar a frequência de cada grupo e `.plot()` para gerar o gráfico em barras.

```python
"""
Aplicamos um filtro isolando o gene TP53. Após a filtragem, utilizamos .value_counts() 
para somar a frequência de cada grupo e .plot() para gerar as barras.
"""
# Filtragem dos dados 
df_TP53 = df[df["Genes"] == "TP53"]

# Contagem dos dados de cada grupo 
contagem_TP53 = df_TP53["Grupo"].value_counts()

# Criação do gráfico
contagem_TP53.plot(kind="bar")
plt.title("Quantidade de Amostras no Grupo Controle")
plt.xlabel("Genes")
plt.ylabel("Número de Pessoas / Amostras")
plt.show()
```

### ✅ **Resolução — Problema 4.3**

**🎯 Objetivo:** Usando uma filtragem simples, faça um histograma comparando a quantidade de amostra de cada grupo de controle para o genes `BRCA1`.

🧠 **Explicação lógica:**
A lógica segue idêntica à do problema anterior, mas agora limitamos a visualização às amostras referentes especificamente à expressividade do gene `BRCA1`.

```python
"""
Lógica idêntica ao problema anterior: limitamos a visualização às amostras referentes
especificamente ao gene BRCA1, contamos os grupos e plotamos.
"""
# Filtragem dos dados 
df_BRCA1 = df[df["Genes"] == "BRCA1"]

# Contagem dos dados de cada grupo 
contagem_BRCA1 = df_BRCA1["Grupo"].value_counts()

# Criação do gráfico
plt.title("Quantidade de Amostras no Grupo Controle")
plt.xlabel("Genes")
plt.ylabel("Número de Pessoas / Amostras")
plt.show()
```

### ✅ **Resolução — Problema 4.4**
**🎯 Objetivo:** Visualize de forma organizada com o objetivo de verificar a relação do grupo de controle com a expressividade dos dois genes.

**OBS:** Este exercício admite muitas formas de criar, seja criativo.

🧠 **Explicação lógica:**
Para criar uma diferenciação visual eficiente, concatenamos a informação do Gene e do Grupo em uma única coluna chamada `Condicao`. Usamos então um gráfico de dispersão (`scatter`) que compara cada grupo lado a lado com seus níveis de expressão.

```python
"""
Criamos uma coluna aglutinando o Gene e o Grupo para ter eixos únicos.
Depois, usamos plt.scatter() para identificar as concentrações de dados no gráfico de dispersão.
"""
df["Condicao"] = df["Genes"] + "_" + df["Grupo"]

plt.scatter(df["Condicao"], df["Expressao"], color="red", alpha=0.6, label="Pacientes")

plt.title("Expressão por Grupo")
plt.suptitle("") 
plt.ylabel("Nível de Expressão")
plt.xlabel("Condições")

plt.show()
```

> ## **Problema 5 - Deslocamento do verme *C. elegans.*** 🟡

Suponha que você está em grupo de pesquisa que está estudando comportamento de vermes da espécie *Caenorhabditis elegans* (verme muito utilizado em pesquisas).
Neste experimento você e seu grupo coletaram as posições relativas de um indíviduo desta espécie, que carinhosamente iremos chamar de `Albert`.  
Dessa forma, neste experimento iremos extrair os dados coletados, manipular esses dados e gerar visualizações.


### ✅ **Resolução — Problema 5.1**

**🎯 Objetivo:** Usando a biblioteca pandas leia o arquivo `dados_rastreamento_albert.csv`, visualize de forma simples a tabela e verifique a integrabilidade dos dados.

🧠 **Explicação lógica:**
Lemos a nova base e usamos o método `.info()`. Esse comando exibe o tipo de dado de cada coluna, o número total de entradas e o número de itens não-nulos, sendo excelente para verificar se perdemos informações ao longo do caminho.

```python
"""
Lemos a nova base e usamos o método .info().
Esse comando é essencial para verificar a presença de itens nulos e as tipagens importadas.
"""
df = pd.read_csv('./dados_rastreamento_albert.csv')

print(df.head())

# Visualização de informações da tabela
df.info()
```

### ✅ **Resolução — Problema 5.2** 

**🎯 Questão:** Com base nas informações obtidas como você avaliaria a integrabilidade dos dados? Quais métricas você utilizou para chegar nesta conclusão?

**Resposta:** Com base nas informações obtidas, podemos afirmar que a integridade dos dados está parcialmente comprometida devido à presença de valores faltantes (nulos). Ao observarmos a saída da célula, notamos a informação `RangeIndex: 100 entries, 0 to 99`, o que indica um total de 100 entradas registradas. Porém, a coluna referente aos dados do eixo x mostra que apenas 98 valores são não-nulos, revelando a existência de 2 dados faltantes. De forma semelhante, a coluna correspondente ao eixo y indica 99 valores não-nulos, apontando para 1 dado faltante nesta coluna.

🧠 **Explicação lógica:**
A análise empírica do log gerado pela função `.info()` comprova diretamente a existência de lacunas numéricas em dados que deveriam ser sequenciais.

### ✅ **Resolução — Problema 5.3**

**🎯 Objetivo:** Com base no exercício anterior, interpole os dados para fazer o tratamento dos dados e verifique novamente.

🧠 **Explicação lógica:**
A função `.interpolate(method='linear')` resolve a falha observada calculando um valor médio que faz a ponte matemática perfeita entre os dados que vêm antes e depois do `NaN` (espaço em branco).

```python
"""
A função .interpolate(method='linear') calcula uma progressão
constante entre os valores adjacentes para cobrir as falhas.
"""
df = df.interpolate(method='linear')

df.info()
```

### ✅ **Resolução — Problema 5.4**

**🎯 Objetivo:** Crie uma visualização bidimenional utilizando as colunas x e y para visualizar o trajeto do `Albert`.

**Nota:** Utiliza a biblioteca `matplotlib.pyplot` e seus métodos para criar a visualização.

🧠 **Explicação lógica:**
Uma simples plotagem linear `.plot(x, y)` usando os dois eixos espaciais recria todo o plano de deslocamento carteseano onde o verme se movimentou na placa.

```python
"""
Uma simples plotagem usando os dois eixos espaciais recria o plano de deslocamento
carteseano com linhas contínuas entre os pontos.
"""
plt.plot(df['x'], df['y'], color='blue', marker='o')

plt.title('Caminho do Albert')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')

plt.grid()

plt.show()
```

### ✅ **Resolução — Problema 5.5**

**🎯 Objetivo:** Crie uma coluna na estrutura de dados que mostra a distância percorrida entre as posições.

🧠 **Explicação lógica:**
Usamos o `.diff()` para deduzir o tamanho da movimentação calculando `ponto_atual - ponto_anterior` de cada eixo, e em seguida extraímos o módulo vetorial dessa distância utilizando pitágoras.

```python
"""
O .diff() nos dá a diferença de deslocamento desde o último ponto.
Através das distâncias absolutas x e y tiramos a reta de distância daquele passo.
"""
df['distancia_passo'] = np.sqrt(df['x'].diff()**2 + df['y'].diff()**2)
```

### ✅ **Resolução — Problema 5.6**

**🎯 Objetivo:** Crie uma visualização para mostrar a distância percorrida entre as posições ao longo do tempo.

🧠 **Explicação lógica:**
Substituindo o plano cartesiano espacial por uma linha temporal em um gráfico de linhas, comparamos os instantes de tempo no eixo horizontal contra o quanto ele andou no eixo vertical.

```python
"""
Comparamos os instantes temporais mapeando o tempo no eixo X e a magnitude calculada do passo no eixo Y.
"""
plt.plot(df['tempo'], df['distancia_passo'], color='green', marker='o')

plt.title('Distância Percorrida por Passo ao Longo do Tempo')
plt.xlabel('Tempo')
plt.ylabel('Tamanho do Passo')

plt.grid()

plt.show()
```
