### **Passo 1.1 — Leitura de .csv**

Como foi dito, as tabelas, acessadas pela biblioteca como DataFrames, são armazenadas no computador em arquivos `.csv`. Então, supondo um arquivo `.csv` a disposição, o nosso primeiro passo é lê-lo.

```python
# =============================================
# PASSO 1.1 — Leitura de .csv
# =============================================

# [PARTE 1] — Trazer a biblioteca para o código
import pandas as pd # Via de regra, Pandas é renomeado como pd, tal qual numpy como np.

# [PARTE 2] — Criar a variável que carregará os dados do arquivo .csv
Tabela = pd.read_csv('exemplo1.csv')
print(Tabela)   # Tente também usar display()!

# Tente ler o arquivo de exemplo 2.
```

> 📝 **Explicação do código acima:**
> - `Tabela` → Variável que recebe o DataFrame
> - `read_csv()` → Transforma o arquivo .csv no `DataFrame` que será apontado pela `Tabela`
> - A entrada da função é o *caminho* do arquivo.

### **Passo 1.2 — Exibição de tabela**
Agora, utilizaremos as funções `set_option()` e `print()` para visualizar o DataFrame armazenado na variável escolhida.

```python
# =============================================
# PASSO 1.2 — Exibição de tabela
# =============================================

# [PARTE 1] — Executar o passo 1
import pandas as pd     # Importa a biblioteca pandas abreviada como "pd"
Tabela = pd.read_csv('exemplo1.csv')    # Faz a leitura de um arquivo como csv

# [PARTE 2] - Visualizar parcialmente
print("PRIMEIRAS LINHAS:")
display(Tabela.head(5))

print("ÚLTIMAS LINHAS:")
display(Tabela.tail(5))

# [PARTE 3] — Formatar a visualização
pd.set_option('display.max_rows', None)     # Determina o limite de linhas (60, por padrão), ou o retira
pd.set_option('display.max_columns', None)  # Análogo ao de cima, mas com um limite de 20 colunas

print("TABELA 'TODA'")
print(Tabela)   # Tente usar display()!

# Imponha limites de linhas e/ou colunas variados.
```

> 📝 **Explicação do código acima:**
> - `'display.max_rows'` → entrada da função que acessa as propriedades do DataFrame, permitindo alterar a configuração desejada.
> - `set_option()` → função setter do objeto Pandas.
> - **Ponto de atenção:** Usar `set_option()` altera as configurações de exibição de todas as exibições seguintes, ou seja, se limitar para 4 uma vez, os próximos prints terão os mesmos 4.

### **Passo 1.3 — Manipulação de linhas e colunas**
Serão selecionadas, fatiadas e alteradas as linhas e colunas a sua escolha do DataFrame. Isso refina o tratamento e análise dos dados, visto a exclusão de ruido ou dados desnecessários no acesso dos dados.

```python
# =============================================
# PASSO 1.3 — Manipulação de linhas e colunas
# =============================================

# Obs: a partir daqui, está suposto a importação e a existência de uma variável Tabela carregando um DataFrame

# [PARTE 1] — Acessar colunas individuais
cidades = Tabela['Cidade']  # Acessa a coluna "Cidade"

display(cidades)
print('\n\n')   # Apenas quebra de linha, ignore

# [PARTE 2] - Acessar linhas individuais
primeira_1 = Tabela.loc[1]
segunda_2 = Tabela.iloc[1]  # Ambos retornam a LINHA 2, porque o Python começa a contar do 0.

display(primeira_1)
print()     # Apenas quebra de linha, ignore
display(segunda_2)
print()     # Apenas quebra de linha, ignore

# [PARTE 3] — Localizar elemento específico
estado_bruno_nome = Tabela.loc[1, ['Estado']]   # Busca pelo NOME da COLUNA
estado_bruno_numero = Tabela.iloc[1,1]          # Busca pelo ÍNDICE da COLUNA

display(estado_bruno_nome)
print()     # Apenas quebra de linha, ignore
display(estado_bruno_numero)
print()     # Apenas quebra de linha, ignore

# [PARTE 4] — Localizar conjuntos de dados
jovens = Tabela[Tabela['Idade'] < 25] # Acessa na tabela todas as linhas em que aquela condição é True
miolo = Tabela.iloc[1:4]                # Pega um intervalo de linhas

display(jovens)
print()     # Apenas quebra de linha, ignore
display(miolo)

# Pegue o elemento central, depois pegue um bloco 9x9 cujo elemento central é o central do DataFrame.
# Pegue pacientes de até 50 anos com a temperatura corporal entre 36.1 e 37 graus.
```

> 📝 **Explicação do código acima:**
> - `.loc[]` → recebe os `endereços` das linhas em listas (ou individual se for apenas um), e recebe os `nomes` das colunas em listas, análogamente.
> - `.iloc[]` → funciona semelhantemente à propriedade `.loc[]`, mas recebe apenas índices numéricos.

### **Passo 1.4 — Cálculo de métricas**
Antes de manipularmos dados complexos, precisamos entender a estrutura geral do nosso DataFrame. O Pandas fornece métodos diretos para inspecionar dimensões, tipos de dados e estatísticas descritivas básicas.

```python
# =============================================
# PASSO 1.4 — Cálculo de métricas e exploração
# =============================================

# [PARTE 1] — Informações gerais e estrutura do DataFrame
Tabela.info()

print(Tabela.shape, '\n\n')     # Exibe a quantidade de linhas e colunas
print(Tabela.columns, '\n\n')   # Agora os nomes das colunas
print(Tabela.index, '\n\n')     # Os índices das linhas
print(Tabela.dtypes, '\n\n')    # Os tipos de dados presentes na tabela

# [PARTE 2] — Estatísticas Descritivas Básicas
print(Tabela.describe(), '\n\n')    # Gera um DataFrame das estatísticas do DataFrame descrito

# [PARTE 3] — Métricas estatísticas isoladas
media_idade = Tabela['Idade'].mean()                    # Média de idades
mediana_temp = Tabela['Temperatura Corporal'].median()  # Mediana das temperaturas
desvio_temp = Tabela['Temperatura Corporal'].std()      # Desvio padrão das temperaturas
soma_idades = Tabela['Idade'].sum()                     # Soma das idades
min_idade = Tabela['Idade'].min()                       # Menor idade
max_idade = Tabela['Idade'].max()                       # Maior idade


print(f"\nMédia de Idade: {media_idade:.2f}")
print(f"Mediana da Temp: {mediana_temp:.2f}")
print(f"Desvio Padrão da Temp: {desvio_temp:.2f}")

# Calcule a dispersão das idades (desvio padrão) e encontre a amplitude da coluna Temperatura Corporal.
```

> 📝 **Explicação do código acima:**
> - `.info()` → Exibe um resumo da estrutura do DataFrame.
> - `.shape` → Retorna o formato `(linhas, colunas)` da tabela.
> - `.describe()` → Gera um resumo numérico das estatísticas da tabela
> - **Ponto de atenção:**  `.shape`, `.columns`, `.dtypes` e `.index` são atributos, e podem ser indexados como uma lista `[1,2,3][0] -> 1`.

### **Passo 1.5 — Busca e contagem de dados**
Determinar as frequências de cada dado numa tabela. Isso permite fazer outras análises estatísticas mais superficiais e permite outras correlacionais.

```python
# =============================================
# PASSO 1.5 — Busca e Contagem de Dados
# =============================================

# [PARTE 1] — Contagem de valores categóricos
contagem_estados = Tabela['Estado'].value_counts()
display(contagem_estados)   # Quantas vezes aquele estado aparece na coluna estado,
                            # ou seja, quantos pacientes de cada estado estão na tabela.
print()

# Veja qual temperatura aparece com maior frequência utilizando.
```

### **Passo 1.6 — Filtragem de dados e agrupamento**
Permite extrair subconjuntos específicos de dados com base em condições lógicas simples ou compostas, e podemos realizar operações agrupadas, como uma análise estatística de uma coluna usando um subconjunto de outra.

```python
# =============================================
# PASSO 1.6 — Filtragem de Dados e Agrupamento
# =============================================

# [PARTE 1] — Filtros com operadores lógicos (& para E, | para OU)
filtro_febre = Tabela[Tabela['Temperatura Corporal'] >= 37.5]                   # Pega todas as temperaturas maiores que 37.5
filtro_complexo = Tabela[(Tabela['Idade'] > 30) & (Tabela['Estado'] == 'RJ')]   # Pega todos os pacientes com idade superior a 30 E
                                                                                # do estado do RJ

# Em ambos os casos, gera-se um DF com as linhas em que o filtro é verdadeiro
display(filtro_febre)
print()
display(filtro_complexo)

# [PARTE 2] — Agrupamento com .groupby()
media_por_estado = Tabela.groupby('Estado')['Temperatura Corporal'].mean()  # Calcula a temperatura média para cada estado
print(media_por_estado) # Faz as médias de temperatura individuais para cada estado

# Gere uma tabela com pacientes hipotérmicos ou acima dos 40 anos.
# Porque precisam ser redirecionados para outra ala de tratamento.
```

Cada caso e cada problema exige métodos específicos para que a tabela possa ser utilizada com o máximo de informação (relevante) possível.

### **Passo 2.1 — Dados faltosos**
Aprenderemos a identificar valores ausentes usando `isnull()` e como removê-los com `dropna()`. Apenas retirar as linhas é uma decisão perigosa na maioria das vezes, visto a perda de outros dados relevantes e que os nulos podem ser substituídos.

```python
# =============================================
# PASSO 2.1 — Tratar Dados Faltosos
# =============================================

# DataFrame com dados a menos

dados_incompletos = {
    'Pacientes': ['Ana', 'Beto', 'Carla', 'Diego', 'Elena'],
    'Idade': [25, None, 30, None, 45],
    'Temperatura Corporal': [36.5, 38.2, None, 36.8, 37.0]
}

tabelinha = pd.DataFrame(dados_incompletos)

# [PARTE 1] — Identificar ausentes

print(tabelinha[tabelinha['Idade'].isnull()])   # Retorna as idades faltando.
print(tabelinha.isnull().sum())                 # Retorna um DF com a quantidade de faltantes em cada coluna.

# [PARTE 2] — Removendo linhas com valores nulos
sem_nulos = tabelinha.dropna()                  # Retira as linhas nulas caso haja ao menos um nela.
sem_sem_idades = tabelinha.dropna(subset=['Idade'])

print(sem_nulos, '\n\n')
print(sem_sem_idades)

# Tente fazer as estatísticas da coluna Idades com e sem os Nans
```

> 📝 **Explicação do código acima:**
> - `.isnull()` → Aplica uma máscara booleana, na qual os dados faltosos são True. Com a função `.notnull()` ocorre o exato oposto.
> - `.shape` → Retorna o formato `(linhas, colunas)` da tabela.
> - `.describe()` → Gera um resumo numérico das estatísticas da tabela
> - **Ponto de atenção:**  `.shape`, `.columns`, `.dtypes` e `.index` são atributos, e podem ser indexados como uma lista `[1,2,3][0] -> 1`.

### **Passo 2.2 — Preenchimento de dados**
Poderemos, daqui para frente, substituir os dados faltosos por algum valor a nossa escolha. Com isso, podemos evitar a retirada de linhas, enviesando ao mínimo os dados.

```python
# =============================================
# PASSO 2.2 — Preenchimento de dados
# =============================================

dados_incompletos = {
    'Pacientes': ['Ana', 'Beto', 'Carla', 'Diego', 'Elena'],
    'Idade': [25, None, 30, None, 45],
    'Temperatura Corporal': [36.5, 38.2, None, 36.8, 37.0]
}
tabelinha = pd.DataFrame(dados_incompletos)

# [PARTE 1] — Preenchimento de vazios números fixos
dados_zeros = tabelinha.copy()                          # Cria uma cópia
dados_zeros['Idade'] = dados_zeros['Idade'].fillna(35)  # Idade média dos brasileiros hoje
                                                        # Podemos preencher os faltosos com um número a nossa escolha.
print(dados_zeros, '\n\n')

# [PARTE 2] — Preenchimento de vazios com estatísticas internas
dados_zeros = tabelinha.copy()
dados_zeros['Idade'] = dados_zeros['Idade'].fillna(
    dados_zeros['Temperatura Corporal'].mean()
)                                                       # O número de preenchimento pode ser a média das nossas amostras. É mais optado no geral.
print(dados_zeros, '\n\n')

dados_zeros['Temperatura Corporal'] = dados_zeros[
    'Temperatura Corporal'
    ].interpolate()                                     # Preenche com a interpolação linear das temperaturas não nulas
print(dados_zeros, '\n\n')
```

> 📝 **Explicação do código acima:**
> - `.copy()` → Cria um outro DataFrame de cópia.
> - `.fillna()` → Preenche os `Nan`s ou `None`s com o argumento da função.
> - **Ponto de atenção:**  Caso apenas faça **dados_zeros = tabelinha**, a tabelinha também será alterada, porque `dados_zeros` aponta para o que será alterado. O `.copy()` faz a variável apontar para uma cópia.

### **Passo 2.3 — Tratamento de datatype, Redução/Renomeação e Salvamento**
Nesta etapa, renomeamos variáveis com `rename()`, ordenamos dados com `sort_values()`, aplicamos transformações personalizadas usando `.apply()`, e salvamos o resultado final em um arquivo `.csv`.

```python
# =============================================
# PASSO 2.3 — Renomeação, Transformação e Salvamento
# =============================================

# [PARTE 1] — Renomeando colunas
Tabela_renomeada = Tabela.rename(
    columns={                               # Passamos um dicionário como parâmetro
        'Pacientes': 'Nome_Paciente',
        'Temperatura Corporal': 'Temp_Corp' # Sempre na estrutura de 'Nome_antigo':'nome_novo'
        })

# [PARTE 2] — Ordenando valores
Tabela_ordenada = Tabela_renomeada.sort_values(
    by='Idade',     # Ordem baseada nas idades
    ascending=False # Maior para a menor (decrescente)
    )
print(Tabela_ordenada, '\n\n')

# [PARTE 3] — Aplicando funções personalizadas com .apply()
def classificar_febre(temp):
    if temp >= 37.8:
        return 'Com Febre'
    else:
        return 'Normal'

Tabela_renomeada[
    'Status_Febre'                  # Como essa coluna não existia, ela é criada
    ] = Tabela_renomeada[
        'Temp_Corp'
        ].apply(classificar_febre)  # Passa a função por toda a coluna desejada

print(Tabela_renomeada, '\n\n')

# [PARTE 4] — Salvando DataFrame em arquivo .csv
Tabela_renomeada.to_csv(
    'Pacientes_Tratados.csv', # Nome do arquivo que será criado
    index=False # False, retira a coluna de índices à esquerda dos dados salvos.
    )

# Faça uma tabela ordenada pela temperatura corporal e salve-a num arquivo novo.
```

Para que os dados e estatísticas das tabelas fiquem mais digeríveis, nós utilizamos as bibliotecas de visualização (seaborn, altair, matplotlib, etc.). Nesse caso, utilizaremos a mais simples (matplotlib) e versátil, porque ela comporta vários tipos de variáveis de entrada, incluindo DataFrames.

### **Passo 3.1 — Gráficos simples**
Agora, aprenderemos os dois tipos mais comuns, simples e versáteis modos de visualizar os dados: os gráficos de barras e os gráficos de dispersão. Além disso, algumas customizações para facilitar o entendimento serão feitas.

```python
# =============================================
# PASSO 3.1 — MatPlotLib: Gráficos Básicos
# =============================================

import matplotlib.pyplot as plt

x = Tabela['Pacientes']
y = Tabela['Temperatura Corporal']
idade = Tabela['Idade']

# [PARTE 1] — Gráfico de Barras com Configurações e Linha Horizontal de Referência
plt.figure(
    figsize=(8, 4) # Matplotlib usa polegadas como unidade de medida do figsize (caso você for imprimir)
    )
plt.bar( # Barras cheias
    x,
    y,
    color='skyblue', # Por padrão é o blue.
    label='Temperatura (°C)' # Rótulo das barras
    )
plt.axhline( # ax(axis, paralelo aos eixos) h(horizontal) line(linha)
    y=37.5,
    color='red',
    linestyle='--', # Indica que a linha é pontilhada
    label='Limite de Febre (37.5°C)' # Nome da linha limite
    ) # Linha de corte/referência


# [PARTE 2] — Customizações
plt.title('Temperatura Corporal dos Pacientes')
plt.xlabel('Pacientes') # Nome do eixo x
plt.ylabel('Temperatura (°C)') # Nome no eixo y
plt.grid(
    True, # Padrão sem a malha
    linestyle=':', # Malha incontínua
    alpha=0.6 # Transparência
    )
plt.legend() # Plota as legendas e rótulos
plt.show()

# [PARTE 3] — Gráfico de Dispersão (Scatter)
plt.figure(figsize=(7, 4))
plt.scatter(
    idade,
    y,
    color='purple',
    s=100 # Tamanho dos pontos
    )
plt.axvline( # Plot de linha vertical no eixo y
    x=40,
    color='orange',
    linestyle='-',
    label='Maior risco de estar acima dos 40 anos'
    )
plt.title('Relação entre Idade e Temperatura Corporal')
plt.xlabel('Idade (anos)')
plt.ylabel('Temperatura Corporal (°C)')
plt.grid(True)
plt.legend()
plt.show()

# Plote um gráfico de linhas (plt.plot) para observar a variação das Idades ao longo dos Pacientes.
```

### **Passo 3.2 — Múltiplos gráficos**
Quando há muitas variáveis, ou mais de um modo que você queira demonstrar a relação entre elas, múltiplos `plots` são necessários.

```python
# =============================================
# PASSO 3.2 — Múltiplos Plots com subplots()
# =============================================

fig, axes = plt.subplots( # Cria uma figura com um grid, cada plot ocupa um espaço desse grid.
    nrows=1,
    ncols=2, # Uma matriz 1x2 de plots
    figsize=(12, 4)
    )

# Plot 1 — Gráfico de Linha
axes[0].plot(
    Tabela['Pacientes'], # Xizes
    Tabela['Idade'], # Ypsilons
    marker='o',
    color='green'
    )
axes[0].set_title('Idade por Paciente')
axes[0].set_xlabel('Pacientes')
axes[0].set_ylabel('Idade')
axes[0].grid(True)

# Plot 2 — Gráfico de Barras
axes[1].bar(Tabela['Pacientes'], Tabela['Temperatura Corporal'], color='coral')
axes[1].set_title('Temperatura por Paciente')
axes[1].set_xlabel('Pacientes')
axes[1].set_ylabel('Temperatura (°C)')
axes[1].grid(True)

plt.tight_layout()
plt.show()

# Crie um subplots com 2 linhas e 1 coluna comparando os mesmos dados.
```

> 📝 **Explicação do código acima:**
> - `.plot()` → Diferente do scatter, essa função junta os pontos por uma linha, necessariamente. Caso você vá plotar o gráfico de uma função, dispondo dos pontos xizes e ypsilons, esta função é a indicada.
> - **Ponto de atenção:**  A função plot liga os pontos na ordem em que foram fornecidos, e são fornecidos em duas entradas (as coordenadas x e as y).
