> ## **Conceito 1 - Trabalhando com arquivos de texto**
No Python, também podemos trabalhar com arquivos externos, como arquivos `.txt`. Veremos o básico de como abrir, ler, modificar e salvar um arquivo `.txt`

Existem três modos básicos:

> **O que vamos aprender:**
> - Criar um arquivo do zero e escrever nele ( write)
> - Ler um arquivo (read)
> - Adicionar informações nova a um arquivo já existente ( append)

Para melhor aproveitar a aula, anexe o arquivo txt enviado como referência

### **Passo 1.1 — Abrindo um arquivo para ler**


Anexe o arquivo "nome_arquivo.txt" no colab para rodar o código. Com a sintaxe abaixo, do `with` é garantido que  o arquivo já será salvo e fechado automaticamente após o encerramento do bloco de código.

```python
# =============================================
# PASSO 1.1 — Abrindo um arquivo para ler
# =============================================

lista_do_arquivo_dos_pacientes = []

with open('nome_arquivo.txt', 'r') as arquivo: # o "r" indica que iremos apenas ler, e não modificar
    for line in arquivo: # para cada linha do meu arquivo
        linha = line.strip()  # Eu vou pegar a linha, porém sem espaço no início e no final
        lista_do_arquivo_dos_pacientes.append(linha) # E vou adicionar na minha lista que criei antes de abrir o arquivo txt

print(lista_do_arquivo_dos_pacientes) # Como o corpo do código já se encerrou, o arquivo já foi fechado, mas eu "salvei" as informações em uma lista
```

> 📝 **Explicação do código acima:**
> * `with open('nome_arquivo.txt', 'r') as arquivo:` → Abre o arquivo especificado no modo de leitura (`'r'`). O uso do `with` é uma boa prática, pois garante que o arquivo seja fechado automaticamente pelo Python assim que o bloco de código terminar, evitando erros e vazamento de memória.
> * `for line in arquivo:` → Percorre o arquivo linha por linha. Essa abordagem é eficiente para arquivos grandes, pois lê o conteúdo sob demanda em vez de carregar tudo na memória de uma vez.
> * `line.strip()` → Remove caracteres invisíveis de quebra de linha (`\n`) e espaços em branco desnecessários nas extremidades de cada linha lida.
> * `lista_do_arquivo_dos_pacientes.append(linha)` → Adiciona o conteúdo limpo de cada linha à lista previamente criada, permitindo que os dados persistam na memória mesmo após o fechamento do arquivo.

### **Passo 1.2 — Criar um arquivo do zero e escrever**


Vamos criar agora o arquivo "pacientes_do_dia.txt" e inserir do zero, todos os pacientes desse dia. Seja criativo e modifique o código para colocar o nome de 2 pacientes de sua escolha nesse arquivo txt novo!

```python
# =============================================
# PASSO 1.2 — Criar um arquivo do zero e escrever
# =============================================

lista_dos_novos_pacientes = ['João Pedro', 'Antônia Maria']

with open('pacientes_do_dia.txt', 'w') as arquivo: # o "w" indica que iremos apenas escrever
    for nome in lista_dos_novos_pacientes: # para cada linha do meu arquivo
        arquivo.write(nome+ '\n') # O +'\n' é necessário para no nosso novo arquivo as linhas serem puladas
```

> 📝 **Explicação do código acima:**
> * `with open('pacientes_do_dia.txt', 'w') as arquivo:` → Abre o arquivo no modo `'w'` (*write*). Atenção: se o arquivo já existir, ele será **sobrescrito** (apagado e criado novamente). O `with` assegura que o arquivo seja salvo e fechado corretamente ao final.
> * `arquivo.write(nome + '\n')` → O método `.write()` insere a string no arquivo. O caractere `'\n'` é fundamental, pois ele instrui o Python a inserir uma **quebra de linha**, garantindo que cada nome seja gravado em uma nova linha no arquivo de texto em vez de ficarem todos grudados.

### **Passo 1.3 — Adicionar informações a um arquivo já existente**


Vamos adicionar ao "nome_arquivo.txt" outros dois nomes. Depois,clique duas vezes no arquivo "nome_arquivo.txt" para verificar que ele foi modificado.

```python
# =============================================
# PASSO 1.3 — Adicionar informações a um arquivo já existente
# =============================================

with open('nome_arquivo.txt', 'a') as arquivo: # o "a" indica que iremos apenas inserir ao final ( append)
    arquivo.write("Andre Barbosa\n")
```

> 📝 **Explicação do código acima:**
> * `with open('nome_arquivo.txt', 'a') as arquivo:` → Abre o arquivo no modo `'a'` (*append*). Diferente do modo `'w'`, este modo não apaga o que já está no arquivo; ele preserva o conteúdo existente e prepara o cursor para escrever **apenas no final** do arquivo.
> * `arquivo.write("Andre Barbosa\n")` → Adiciona a nova informação ao final do arquivo. O uso do `\n` continua sendo indispensável para garantir que essa nova entrada inicie em uma linha própria, sem ficar colada no conteúdo que já existia anteriormente.
