> ## **Conceito 1 - Listas**

### **Passo 1.1 — Criando e exibindo listas**
Para criar uma lista, precisamos utilizar **colchetes `[ ]`** e escrever seus elementos, todos separados por **vírgula**

```python
# =============================================
# PASSO 1.1 — Criando e exibindo listas
# =============================================

# [PARTE 1] — Criando uma lista
lista_1 = []                            # Uma lista vazia
lista_2 = [1,2,3]                       # Uma lista com inteiros
lista_3 = [1.5 , 9,2 , 3,7]             # Uma lista com floats
lista_4 = ["Elemento 1", "Elemento 2"]  # Uma lista com strings

# [PARTE 2] — Misturando tipos
# Os elementos de uma lista podem ter QUALQUER TIPO

lista_5 = [1,2.5,"String",True] # Uma lista com int, float, string, bool
lista_6 = [[1,2,3],[4,5,6]]     # Uma lista contendo outras duas listas

# [PARTE 3] - Otimizações
# Para facilitar a visualização, podemos também usar espaços e quebras de linha para criar listas

lista_7 = [1.1 , 9.5 , 3.5] # Podemos adicionar espaços ao redor das vírgulas, facilitando a visualização
lista_8 = [[1,2,3],         # Também podemos usar quebras de linha após a vírgula
           [4,5,6],
           [7,8,9]]

# [PARTE 4] - Exibindo listas
print(lista_3)
print(lista_5)
print(lista_6)

#  AGORA É SUA VEZ: tente criar uma lista e exibir na tela
```

### **Passo 1.2 — acessando elementos em uma lista**
Cada elemento da lista possui um **índice** (posição), iniciando em **0**. Podemos acessar um elemento da lista usando `lista[indice]` <br>

Digamos que eu precise acompanhar o índice de massa de alguns pacientes, e para esse estudo, estamos considerando que as únicas informações que estão sendo estudadas no momento é sexo, idade e o índice de massa corporal. Dessa forma, foi estabelecia uma convenção onde todo paciente teria uma lista (ficha) onde a primeira informação é o seu gênero, a segunda a sua idade e a terceira, armazenaríamos o índice.

```python
# =============================================
# PASSO 1.2 — Acessando elementos em uma lista
# =============================================

# [PARTE 1] — Acessando por índice
ficha1 = ["Homem",  # Sexo
          37,       # Idade
          0.2]      # Índice

indice_massa = ficha1[2]
sexo = ficha1[0]

print(indice_massa) # Retorna o IMC
print(ficha1[2])    # Que é o elemento com índice 2 na lista

print(sexo)         # Retorna o sexo
print(ficha1[0])    # Que é o elemento com índice 0 na lista

# [PARTE 2] - Índice negativos
# Caso tenhamos uma lista muito extensa, também podemos acessar elementos
# utilizando índices negativos (nesse caso, a contagem é feita "do último para o primeiro")
indice_massa = ficha1[-1]
sexo = ficha1[-3]

print(indice_massa) # Retorna o IMC
print(sexo)         # Retorna o sexo
```

> 📝 Explicação do código acima:
> - `ficha1[0]` e `ficha1[2]` → A contagem das posições em uma estrutura de dados sempre começa do zero. Por isso, o índice `0` acessa o primeiro elemento da lista (`"Homem"`), e o índice `2` acessa o terceiro elemento (`0.2`).
> - `ficha1[-1]` e `ficha1[-3]` → O índice `-1` é um atalho extremamente útil que sempre retorna o **último** elemento da lista (neste caso, o IMC `0.2`), sem que você precise saber o tamanho total dela. Seguindo a lógica, o `-3` volta três posições a partir do final, acessando o primeiro item (`"Homem"`).

### **Passo 1.3 — Adicionando e removendo elementos**
As principais funções que utilizamos para adiciomar e remover elementos de uma lista são:

| Função     | Parâmetros                       | O que faz                                                                          | Observações                                                                           |
| :--------- | :------------------------------- | :--------------------------------------------------------------------------------- | :------------------------------------------------------------------------------------ |
| `append()` | `elemento`                       | Adiciona um elemento ao final da lista.                                            | O novo elemento ocupa o último índice da lista.                                       |
| `insert()` | `índice, elemento`               | Adiciona um elemento em uma posição específica.                                    | Os elementos a partir da posição informada são deslocados uma posição para a direita. |
| `extend()` | `iterável`                       | Adiciona todos os elementos de outra lista                                         | Cada elemento é adicionado individualmente ao final da lista.                         |
| `remove()` | `elemento`                       | Remove a primeira ocorrência do elemento informado.                                | -                                                                                     |
| `pop()`    | `índice` *(opcional)*            | Remove **e retorna** o elemento da posição informada.                                | Se nenhum índice for informado, remove o último elemento da lista.                    |
| `clear()`  | *Nenhum*                         | Remove todos os elementos da lista.                                                | A lista continua existindo, mas fica vazia (`[]`).                                    |

```python
# =============================================
# PASSO 1.3 — Adicionando e removendo elementos
# =============================================

lista = [1,2,1,4,8,6]
lista_aux = [10,11,12]
print(lista)

# [PARTE 1] — Adicionando elementos
lista.append(1)                 # Adiciona um elemento int(1)
lista.append("Item")            # Adiciona um elemento str("Item")
print(lista)

lista.insert(0, "Novo item")    # Insere o elemento no índice 0
print(lista)

lista.extend(lista_aux)         # Adiciona os elementos da lista_aux

# [PARTE 2] - Removendo elementos
lista.remove(1)                 # Remove a primeira aparição de int(1) da lista (que está no índice 1)
print(lista)

print(lista.pop())              # Remove e RETORNA o último elemento da lista
print(lista.pop(0))             # Remove e RETORNA o elemento de índice 0 da lista
print(lista)                    # O último elemento (12) foi removido na linha acima

lista[0] = "Item modificado"    # Também podemos modificar elementos diretamente, indicando seu índice
print(lista)

lista.clear()                   # Remove todos os elementos da lista
print(lista)
```

> 📝 Explicação do código acima:
> - `lista.append(1)` → Adiciona o elemento `1` na lista. O elemento é adicionado na última posição a direita. O mesmo ocorre para `lista.append("Item")`
> - `lista.insert(0,"Novo item")` → Adiciona o elemento "Novo item" no índice 0 (primeira posição) da lista. Note que todos os outros tem seus índices "movidos" para a direita
> - `lista.extend(lista_aux)` → Adiciona cada elemento de `lista_aux` na `lista`. Como fazer um `.append()` em cada item de `lista_aux`
> - `lista.remove(1)` → Remove a primeira aparição de `1` na lista. Note que, mesmo que tenham outros `1`'s na lista, a função remove apenas o primeiro.
> - `print(lista.pop())` → O trecho `lista.pop()` remove o último item da lista, e o **retorna**. É como "ler" qual o último item e removê-lo em seguida. Usamos o `print()` para conseguir ver qual esse item
> - `print(lista.pop(0))` → Faz exatamente o mesmo que o comando acima, mas, indicamos o  elemento do índice `0` para ser removido (e retornado).
> - `lista[0] = "Item modificado` → Altera diretamente o item de índice `0` da lista, para o valor `Item modificado`.
> - `lista.clear()` → Remove todos os elementos da lista. A lista **não** deixa de existir, ela apenas se torna uma lista **vazia**.

### **Passo 1.4 — Fatiamento de listas**
Podemos usar o fatiamento para "cortar" a lista em outras sub-listas. A lógica usada vai ser padrão para a maioria dos iteráveis do Python (como strings, tuplas e dicionários, que serão vistos a frente). Para isso, indicamos um intervalo usando `lista[inicio:fim]`

```python
# =============================================
# PASSO 1.4 — Fatiamento de listas
# =============================================

lista_original = ['a', 'b', 'c', 'd', 'e', 'f', 'g']

# [PARTE 1] - Intervalos fechados
# O limite inferior é incluído, mas o superior não.
# Retorna uma lista contendo os elementos a partir do índice "1" até o índice "3"
fatiamento1 = lista_original[1:4]

print("fatiamento 1 =", fatiamento1)
print("lista original se mantém como:", lista_original)
print()

# [PARTE 2] - Intervalos abertos
# Quando não indicamos um dos limites do intervalo, o Python usa como padrão
# o índice 0 (1º elemento) ou o último índice (último elemento)

fatiamento2 = lista_original[3:]  # Lista que começa no índice 3 e vai até o final
print("fatiamento 2 =", fatiamento2)

fatiamento3 = lista_original[:3]  # Lista do início até o índice 3 (limite não incluso)
print("fatiamento 3 =", fatiamento3)

fatiamento4 = lista_original[:]   # Cópia idêntica da lista original
print("fatiamento 4 (copia da lista) =", fatiamento4)
print()

# [PARTE 3] - Alterando a lista original com fatiamento
# Podemos usar o fatiamento para substituir trechos da lista

lista_original[1:3] = ['j', 'k']
print("Substituição exata:", lista_original)

# O que aconteceria se colocássemos mais elementos do que o permitido pelo intervalo?
# O Python empurra os demais itens para a direita, aumentando a lista
lista_original[1:3] = ['r', 's', 't', 'u']
print("Inserindo mais elementos:", lista_original)

# E se colocássemos menos?
# O Python encolhe a lista, removendo os elementos que "sobraram" na substituição
lista_original[1:4] = ['p']
print("Inserindo menos elementos:", lista_original)
print()

# [PARTE 4] - Concatenando listas
# Também podemos concatenar (somar/juntar) listas diferentes

UFC_segunda = [100000, 50000, 70000, 300000, 40000]
UFC_terca = [200000, 30000, 10000, 900000, 9000]
UFC_quarta = [800000, 15000, 45000, 89000, 40000]
UFC_quinta = [50000, 60000, 70000, 200500, 44000]
UFC_sexta = [123000, 58900, 80000, 356000, 45900]

UFC_total = UFC_segunda + UFC_terca + UFC_quarta + UFC_quinta + UFC_sexta

print("UFC_total =", UFC_total)
```

> 📝 Explicação do código acima:
> - `lista_original[1:4]` → Extrai um "pedaço" da lista. O primeiro número (1) é o índice onde o corte começa (inclusivo). O segundo número (4) é onde o corte termina, mas ele **não é incluído**. Ou seja, pega os índices 1, 2 e 3.
> - `lista_original[3:]` e `lista_original[:3]` → Se você não colocar o número antes dos dois pontos, o Python assume que é o começo da lista (índice 0). Se não colocar o número depois, ele assume que deve ir até o final da lista.
> - `lista_original[:]` → Ao não indicar nenhum dos dois limites, o Python pega do primeiro ao último elemento (padrão).
> - `lista_original[1:3] = [...]` → O Python permite trocar um trecho inteiro da lista de uma vez. Se você inserir mais elementos do que o tamanho do espaço, a lista cresce. Se inserir menos, a lista encolhe automaticamente.
> - `UFC_segunda + UFC_terca + ...` → O operador de soma (`+`), quando usado com listas, simplesmente "cola" uma lista no final da outra, criando uma lista nova e maior contendo todos os elementos.

### **Passo 1.5 — Outras funções**
Algumas outras funções que podemos usar em listas são:

| Função      | Parâmetros                   | O que faz                                                     | Observações                                                                                                                |
| :---------- | :--------------------------- | :------------------------------------------------------------ | :------------------------------------------------------------------------------------------------------------------------- |
| `index()`   | `elemento`                   | Retorna o índice da primeira ocorrência do elemento na lista. | -                                                           |
| `count()`   | `elemento`                   | -                                                             | -                                                               |
| `sort()`    | `reverse=False` *(opcional)* | Ordena os elementos da lista.                                 | Modifica a própria lista. Por padrão, ordena em ordem crescente.                                                           |
| `reverse()` | *Nenhum*                     | Inverte a ordem dos elementos da lista.                       | Apenas inverte sua ordem atual.                                                                        |
| `len()`     | `lista`                      | Retorna a quantidade de elementos da lista.                   | -                                                                |
| `max()` e `min()`     | `lista`                      | Retorna o maior ou o menor elemento da lista, respectivamente.                            | Os elementos devem ser comparáveis entre si (como números, por exemplo).                                                                               |                                                                       
| `sum()`     | `lista`                      | Retorna a soma dos elementos da lista.                        | -                                                       |
| `in`        | `elemento in lista`          | Verifica se um elemento pertence à lista.                     | Retorna `True` se o elemento estiver na lista e `False` caso contrário. `in` é um operador da linguagem, e não uma função. |


```python
# =============================================
# PASSO 1.5 — Outras funções e métodos
# =============================================

lista = [0,4,2,3,4,1,5,6,7,8,1,9]

print(lista.index(4)) # Retorna 1

print(lista.count(1)) # Retorna 2

lista.sort()          # Ordena a lista em ordem crescente (podemos usar o reverse=False)
print(lista)          # caso queiramos em ordem decrescente

lista.reverse()       # Inverte a lista
print(lista)          # Exibe a lista invertida

print(len(lista))     # Retorna 12

print(max(lista))     # Retorna 9
print(min(lista))     # Retorna 0

print(sum(lista))     # Retorna 47

print(5 in lista)     # Retorna True
print("Olá" in lista) # Retorna False
```

> 📝 Explicação do código acima:
> - `lista.index(4)` → Retorna qual a posição do **primeiro** valor 4 dentro da lista, que é a segunda posição (índice 1). Caso o elemento fornecido não exista na lista, retorna um erro
> - `lista.count(1)` → Retorna a quantidade de valores iguais a `1` dentro da lista. Qualquer valor pode ser fornecido como parâmetro.
> - `lista.sort()` → Ordena os elementos da lista, do menor para o maior (valores iguais aparecem em sequência). Perceba que, isso modifica diretamente a variável `lista`, como pode ser visto no `print(lista)` logo abaixo
> - `lista.reverse()` → Semelhante ao `sort()`, mas essa função inverte a ordem dos elementos. Isto é, o último elemento da lista se torna o primeiro, o penúltimo se torna o segundo, e assim por diante. A função `reverse()` **NÃO** ordena a lista do maior para o menor, isso aconteceu apenas por que usamos o `sort()` logo antes. Da mesma forma, essa função também altera diretamente a variável `lista`
> - `len(lista)` → Retorna a quantidade de elementos da lista
> - `max(lista)` e `min(lista)` → Retorna o maior e o menor valor da lista, respectivamente
> - `sum(lista)` → Retorna a soma de todos os valores da lista
> - `5 in lista` → Como a lista contém um elemento `5`, retorna `True`
> - `"Olá" in lista` → Como a lista **NÃO** contém nenhum elemento `"Olá"`, retorna `False`

> ## **Conceito 2 - Tuplas**
As tuplas são muito semelhantes as listas, porém, sua principal diferença é que, tuplas são **imutáveis**. Isto é, uma vez que ela foi definida, ela **não pode ser modificada**. Ou seja: não podemos adicionar, remover ou modificar seus elementos.

### **Passo 2.1 — Criando e exibindo tuplas**
Para criar uma tupla, precisamos utilizar **parênteses `( )`** e escrever seus elementos, todos separados por **vírgula**

```python
# =============================================
# PASSO 2.1 — Criando e exibindo tuplas
# =============================================

# [PARTE 1] - Criando uma tupla
tupla = (1,2,3,4)

# [PARTE 2] - Tentando modificar

tupla[0] = 2  # Se tentamos modificar um de seus elementos, temos um ERROR
```

```python
# =============================================
# PASSO 2.1 — Criando e exibindo tuplas
# =============================================

# [PARTE 3] - "Mudando" uma tupla
tupla = (1,2,3,4)
tupla = (1,2,3)

print(tupla)
```

> 📝 Explicação do código acima:
> - Nós não modificamos a tupla `(1,2,3,4)`, nós apenas atribuímos uma tupla diferente `(1,2,3)` a variável `tupla`

### **Passo 2.2 — Métodos com tuplas**
A maneira como trabalhamos com tuplas é muito semelhante a maneira como trabalhamos com listas, diferindo-se de toda e qualquer método que envolva alguma **modificação na tupla**

```python
# =============================================
# PASSO 2.2 — Métodos com tuplas
# =============================================

tupla = (0,4,2,3,4,1,5,6,7,8,1,9)

# Podemos acessar elementos específicos
print(tupla[1])

# Podemos fatiar a tupla
print(tupla[2:5])
print(tupla[:5])
print(tupla[5:])

# Podemos saber qual o índice de algum elemento
print(tupla.index(4))

# Podemos contar quantas aparições um determinado elemento tem
print(tupla.count(1))

# Podemos saber quantos elementos tem na tupla
print(len(tupla))

# Podemos obter seu máximo e mínimo
print(max(tupla))
print(min(tupla))

# Podemos calcular a soma de seus elementos
print(sum(tupla))

# Podemos ver se um item pertence a tupla
print(5 in tupla)
print("Olá" in tupla)
```

> ## **Conceito 3 - Dicionários**
Uma **lista** é uma sequência de elementos (ou itens) em sequência. Semelhante a uma string, que é uma sequência de caracteres, uma lista é uma sequência de itens que podem ser de **qualquer tipo**

> **O que vamos aprender:**
> - Criando e exibindo um dicionário
> - Acessando elementos em um dicionário
> - Como adicionar e remover elementos

### **Passo 3.1 — Criando e exibindo um dicionário**
Para criar um dicionário, devemos:
- Digitar uma **chave**, que pode ter **qualquer** valor (um número, um texto, etc.)
- Usar `:` para ligar cada **chave** a seu respectivo **valor**
- Separar os pares de chaves e valores por vírgula
- Colocar tudo isso entre chaves `{ }`

Na prática, teremos algo como: <br><br> `dicionario = {chave1 : valor 1, chave2 : valor 2}`

```python
# =============================================
# PASSO 3.1 — Criando e exibindo dicionários
# =============================================

# [PARTE 1] — Criando um dicionário
dicionario_1 = {}                                         # Um dicionário vazio
dicionario_2 = {"chave_1":1, "chave_2":2, "chave_3":3}    # Um exemplo de dicionário

# [PARTE 2] — Misturando tipos
# Os elementos de uma lista podem ter QUALQUER TIPO

dicionario_3 = {1 : True, "Dict" : [1,2]}                 # Tanto as chaves quanto os valores podem ter qualquer tipo
dicionario_4 = {"dic1" : {1:1, 2:2}, "dic2" : {3:3, 4:4}} # Inclusive outros dicionários

# [PARTE 3] - Otimizações
# Da mesma forma que listas, também podemos utilizar quebras de linha e espaços para facilitar a visualização

dicionario = {"chave_1" : "valor_1",
              "chave_2" : "valor_2 ",
              "chave_3" : "valor_3 ",
              "chave_4" : "valor_4 ",}

# [PARTE 4] - Exibindo dicionários
print(dicionario_1)
print(dicionario_2)
print(dicionario)

# Perceba que, é muito fácil que a exibição de dicionários em uma linha torne-se confusa
# A função display() ajuda a visualizarmos mais claramente

print()             # Apenas uma quebra de linha
display(dicionario) # Exibe o dicionário de forma mais legível
```

---

### **Passo 3.2 — Acessando elementos em um dicionário**

O processo para acessar elementos em um dicionário é semelhante ao que usamos para acessar elementos em uma lista, porém, ao invés de índices, devemos usar suas **chaves**.

```python
# =============================================
# PASSO 3.2 — Acessando elementos em um dicionário
# =============================================

# [PARTE 1] - Acessando por chave
paciente = {"nome":"Ryan Kevin",
            "idade":20,
            "profissao":"Estudante",
            "estado_grave":False,
            10:5}

print(paciente["nome"])                   # Acessa o valor de "nome", que é "Ryan Kevin"
print(paciente["idade"])                  # Acessa o valor da chave "idade", que é 20
print(paciente[10])                       # Acessa o valor da chave 10, que é 5

# [PARTE 2] - Método get()

print(paciente.get("nome"))               # Acessa o valor da chave "nome"
print(paciente.get("Data de nascimento")) # Caso a chave não exista, retorna None

# [PARTE 3] - Lista de chaves e lista de valores

chaves = list(paciente.keys())            # Retorna uma lista com as chaves do dicionário
print(chaves)

valores = list(paciente.values())         # Retorna uma lista com os valores do dicionário
print(valores)

print(chaves[0], valores[0])

# [PARTE 4] - Lista de tuplas (chave, valor)

items = list(paciente.items())            # Retorna uma lista com tuplas (chave, valor) do dicionário
print(items)

print(items[0])
```

> 📝 Explicação do código acima:
> - `paciente["nome"]` → acessa diretamente o valor associado à chave. Repare que as chaves podem ser tanto textos (`"idade"`) quanto números (`10`), quanto qualquer outro tipo de variável.
> - `paciente.get("chave")` → Diferente do acesso por colchetes, se você buscar uma chave que não existe (como `"Data de nascimento"`), o `.get()` não quebra o programa com um erro.
> - `list(paciente.keys())` e `list(paciente.values())` → extraem as chaves e os valores separadamente. O uso do comando `list()` em volta serve para converter o resultado visual em uma **lista tradicional**, permitindo que você acesse os itens por índice.
> - `list(paciente.items())` → Semelhante os dois trechos acima, porém, retorna uma uma lista de **tuplas** no formato `(chave, valor)`.

### **Passo 3.3 — Adicionando e removendo elementos**

Para adicionar ou remover elementos, podemos utilizar alguns métodos:

```python
# =============================================
# PASSO 3.3 — Adicionando e removendo elementos
# =============================================

paciente = {"nome":"Ryan Kevin",
            "idade":20,
            "profissao":"Estudante",
            "estado_grave":False,}

# [PARTE 1] - Adicionando itens

paciente["sintomas"] = ["Tosse", "Dor de cabeça"] # Se a chave "sintomas" não existe, ela é criada
paciente["profissao"] = "médico"                  # O valor da chave "profissao" é atualizado
paciente.update({"altura":1.74, "peso":76.2})     # Adicionamos as chaves "altura" e "peso"
display(paciente)

# [PARTE 2] - Removendo itens

print()
print(paciente.pop("profissao"))  # Remove a chave "profissao" do dicionário e imprime o valor removido ("médico")
print()

display(paciente)
paciente.clear()                  # Limpa o dicionário inteiro, removendo todos os itens de uma vez

print()
display(paciente)                 # Exibe o dicionário vazio: {}
```

> 📝 Explicação do código acima:
> - `paciente["sintomas"] = [...]` e `paciente["profissao"] = "médico"` → Se a chave não existir no dicionário (como `"sintomas"`), ela é criada. Se já existir (como `"profissao"`), o valor antigo é substituído pelo novo.
> - `paciente.update(...)` → Podemos adicionar vários pares de chave e valor simultaneamente, bastando passar um novo dicionário como parâmetro.
> - `paciente.pop("profissao")` → Funciona identicamente ao método `.pop()` das listas: a chave é removida e retornada
> - `paciente.clear()` → Apaga todas as chaves do dicionário, transformando-o em um dicionário vazio `{ }`

### **Passo 4.1 — Iterando listas e tuplas**

No geral, usaremos `for ... in ...` para **iterar** sobre estruturas de dados

```python
# =============================================
# PASSO 4.1 — Iterando listas e tuplas
# =============================================

lista = ["Ryan", "Cristiane", "Bianca"]
tupla = ("Paracetamol", "Omeprazol", "Losartana")

for pessoa in lista:  # Lembrando que podemos dar qualquer nome a variável de controle
  print(pessoa)

for remedio in tupla:
  print(remedio)
```

### **Passo 4.2 — Iterando dicionários**

Para iterar sobre dicionários, temos três possibilidades:
- Iterar sobre as **chaves** do dicionário
- Iterar sobre os **valores** do dicionário
- Iterar sobre os pares **chave e valor** do dicionário

Para isso, utilizaremos os métodos `keys()`, `values()` e `items()` que retornam uma **lista** dos dados que estão no dicionário, como visto anteriormente. Então, usamos a iteração sobre listas nestas listas

```python
# =============================================
# PASSO 4.2 — Iterando dicionários
# =============================================

# Imagine um dicionário que guarda as temperaturas de pacientes
dicionario = {"Ryan Kevin":37.5,
              "Leo Bastos":38.1,
              "Carlos Eduardo":36.9,
              "Cristiane Sampaio":38.9}

# [PARTE 1] - Iterando sobre chaves
# Iterando nas chaves, podemos analisar os nomes dos pacientes que estão nos nossos dados

for paciente in dicionario.keys():
  print(paciente)

  # [PARTE 2] - Iterando sobre valores
  # Iterando nos valores, podemos analisar as temperaturas da nossa amostra de pacientes

for temperatura in dicionario.values():
  print(temperatura)

  # [PARTE 3] - Iterando sobre chaves e valores
  # Iterando em chaves e valores, podemos saber qual a temperatura de cada paciente da amostra

for paciente, temperatura in dicionario.items():
  print(paciente,temperatura)
```

> 📝 Explicação do código acima:
> - `for paciente in dicionario.keys():` → O loop `for` percorre apenas a lista de chaves do dicionário. A cada volta do loop, a variável `paciente` recebe um dos nomes (ex: "Ryan Kevin", "Leo Bastos") e o imprime na tela.
> - `for temperatura in dicionario.values():` → O loop ignora quem são os pacientes e foca exclusivamente na lista de valores. A variável `temperatura` vai assumindo as medições (37.5, 38.1, etc.) a cada repetição.
> - `for paciente, temperatura in dicionario.items():` → O método `.items()`, como foi visto, retorna uma **lista de tuplas** `(chave, valor)`, o Python já desempacota isso automaticamente: a primeira variável (`paciente`) recebe a chave, e a segunda (`temperatura`) recebe o valor associado a ela.

### **Passo 5.1 — Fatiamento de strings**
Podemos usar o fatiamento para "cortar" a string em outras sub-strings.Como dito anteriormente, utiliza a mesma lógica de listas. Indicamos um intervalo usando `string[inicio:fim]`

```python
# =============================================
# PASSO 5.1 — Fatiamento de strings
# =============================================

string_original = 'Paracetamol'

# [PARTE 1] - Intervalos fechados
# O limite inferior é incluído, mas o superior não.
# Retorna uma lista contendo os elementos a partir do índice "1" até o índice "3"
fatiamento1 = string_original[1:4]

print("fatiamento 1 =", fatiamento1)
print("string original se mantém como:", string_original)
print()

# [PARTE 2] - Intervalos abertos
# Quando não indicamos um dos limites do intervalo, o Python usa como padrão
# o índice 0 (1º elemento) ou o último índice (último elemento)

fatiamento2 = string_original[3:]  # Lista que começa no índice 3 e vai até o final
print("fatiamento 2 =", fatiamento2)

fatiamento3 = string_original[:3]  # Lista do início até o índice 3 (limite não incluso)
print("fatiamento 3 =", fatiamento3)

fatiamento4 = string_original[:]   # Cópia idêntica da lista original
print("fatiamento 4 (copia da lista) =", fatiamento4)
print()

# [PARTE 3] - Alterando a string original
# Atenção! Diferentemente de uma lista, em strings NÃO podemos usar o fatiamento para substituir trechos da lista
# Isso ocorre pois strings são imutáveis. O que pode ser feito para ter a string modificada é:

#Opção 1: Criar uma string nova com fatiamento:

nova_string = string_original[:1] + "j k" + string_original[3:]

print("Nova string modificada :",nova_string)

#Opção 2: Criar nova string com método replace(), com parametros _elemento_que_será_substituido  e _novo_elemento
#O método replace() consegue tirar caracteres através da substituição por uma string vazia.
#replace() realiza uma substituição. Logo, também viabiliza uma inserção de elemento no meio da string com uma substituição esperta. Olhe:

print('string original =',string_original)

string_replace = string_original.replace('ar','j k')

print("Substituição na cópia com replace:", string_replace)


# E se colocássemos menos elementos no replace?

string_replace = string_original.replace('j k','')

print("Inserindo menos elementos:", string_replace)
print()

# [PARTE 4] - Concatenando strings
# Também podemos concatenar (somar/juntar) strings diferentes

palavra_1 = 'Gostaria '
palavra_2 = 'de '
palavra_3 = 'participar '
palavra_4 = 'do '
palavra_5 = 'estudo.'

frase = palavra_1 +  palavra_2 +  palavra_3 +  palavra_4 +  palavra_5

print("Frase total =", frase)
```

> 📝 Explicação do código acima:
> - `string_original[1:4]` → Funciona exatamente como nas listas, extraindo um trecho da *string*. O índice de início é inclusivo, mas o de parada não.
> - `string_original[3:]` e `[:3]` → Seguem a mesma lógica de "atalho" das listas. Omitir o valor antes dos dois pontos indica o início da *string*, e omitir depois indica o fim.
> - **Imutabilidade:** Diferente das listas, você **não pode** alterar uma *string* diretamente usando `string[1:3] = 'novo'`. Strings são imutáveis em Python.
> - `nova_string = string_original[:1] + "j k" + ...` → Como não podemos alterar a original, a solução é "montar" uma nova *string* unindo partes fatiadas com novos conteúdos usando o operador `+`.
> - `string_original.replace('ar', 'j k')` → Este método busca um padrão e substitui por outro, gerando uma cópia modificada.
> - `palavra_1 + palavra_2 + ...` → O operador `+` com *strings* funciona como uma "cola", unindo os textos em sequência para formar uma nova *string* completa (frase).

### **Passo 5.2 — Contagem**
Quando falamos em "contagem" em uma string podemos estar falando tanto da função que retorna quantos elementos ao todo a string tem, ou quantas ocorrências existem de uma substring nela. Com isso, temos:

> - Tamanho string
> - Contagem de substrings

```python
# =============================================
# PASSO 5.2 — Contagem
# =============================================

# [PARTE 1] Retornando tamanho da string

frase = 'a semana tem sete dias'

tamanho = len(frase) # Observação: conta com os espaços

print('tamanho da frase contando com os espaços =',tamanho)

# [PARTE 2] Contagem de substrings

ocorrencias_de_a = frase.count('a')

print('ocorrencias da substring a =',ocorrencias_de_a)
```

> 📝 Explicação do código acima:
> - `len(frase)` → Retorna a quantidade total de caracteres presentes na *string*. É importante notar que, em Python, o espaço em branco é considerado um caractere, portanto, ele é incluído na contagem final.
> - `frase.count('a')` → Analisa a *string* e retorna quantas vezes a sequência de caracteres especificada (neste caso, a letra 'a') aparece dentro dela.

### **Passo 5.3 — Separação por Espaço**
Um método muito utiplzado é o split(). Isso ocorre pois não necessariamente uma entrada estará formatada da forma mais conveniente ( que é cada elemento sendo enviado em linhas diferentes, como os exemplos numéricos vistos nas últimas aulas). Ou então, temos um problema onde sabemos que teremos que atender n pacientes, porém ao invés da entrada ter uma linha por nome de paciente, recebemos apenas uma linha, com todos os nomes separados por espaço. Por isso o split() é importante. Ele transforma uma string em uma lista, cujos os elemenos nas diferentes entradas das listas são separadas pelo espaço. Observe:

```python
# =============================================
# PASSO 5.3 — Separação por Espaço
# =============================================

dado_da_entrada = 'João Marcos André Joana Márcia Antônia'

lista_pacientes = dado_da_entrada.split()

print(lista_pacientes)
```

> 📝 Explicação do código acima:
> - `dado_da_entrada.split()` → Este método analisa a *string* e quebra o conteúdo onde encontrar espaços em branco. Como resultado, ele transforma a *string* original em uma **lista**, onde cada palavra passa a ser um elemento individual (neste caso, a lista de nomes). Se nenhum caractere for especificado dentro dos parênteses, o Python assume que o delimitador padrão é qualquer espaço em branco.

A função split() tem o parâmetro do que você deseja usar como referência de separador. Coloque como entrada a seguinte string, e veja o que o código retorna:

`Cristiane Sampaio, Carlos Eduardo, Ryan Kevin`

```python
# =============================================
# PASSO 5.3 — Separação por Espaço (input)
# =============================================

frase_entrada = input('Coloque a entrada aqui:')

lista_pacientes = frase_entrada.split(', ')

print(lista_pacientes)
```

> 📝 Explicação do código acima:
> - `frase_entrada.split(', ')` → Diferente do `.split()` simples, aqui passamos um argumento (`', '`). Isso instrui o Python a quebrar a *string* especificamente onde houver uma vírgula seguida de um espaço, ignorando outros espaços que possam existir.

### **Passo 5.4 — Manipulação**


Vamos ver agora uma tabela das diferentes funções de string e depois um exemplo de utilização. Lembre sempre que strings são imutáveis, então com exceção da concatenação vista anteriormente, para qualquer uma dessas modificações serem salvas, elas tem que estar atribuídas a uma string nova.


| Função      | Parâmetros                   | O que faz                                                     | Observações                                                                                                                |
| :---------- | :--------------------------- | :------------------------------------------------------------ | :------------------------------------------------------------------------------------------------------------------------- |
| `find()`   | `substring`                   | Retorna o índice da primeira ocorrência da substring, ou -1 caso não exista. | -                                                           |
| `count()`   | `substring`                   | Retorna quantas vezes a substring ocorre | -                                                                                               |
| `strip()`    | `elemento` *(opcional)* | Remove o `elemento`das extremidades ( se houver). | Remove espaço das extremidades se nenhum `elemento` for passado.                                 | Modifica a própria lista. Por padrão, ordena em ordem crescente.                                                           |
| `upper()` | *Nenhum*                     | Converte a string para letras maíusculas                       | _                                                                        |
| `lower()` | *Nenhum*                     | Converte a string para letras minúsculas                       | _                                                                        |
| `capitalize()` | *Nenhum*                     | Converte a string para  que a primeira letra seja maiúscula                     | _                                                                        |
| `title()` | *Nenhum*                     | Converte a string para que toda primeira letra seja maiúscula                      | _                                                                        |
| `len()`     | `lista`                      | Retorna a quantidade de elementos da string.                   |
| `in`        | `elemento in string`          | Verifica se um elemento pertence à string.                     | Retorna `True` se o elemento estiver na string e `False` caso contrário. `in` é um operador da linguagem, e não uma função. |-                                                                |
| `isdigit()`        |  -        | Verifica se a string contém apenas dígitos.                     | Retorna `True` se só tiver dígitos e `False` caso contrário.  |-                                                                |
| `isalpha()`        | -         | Verifica se a string contém apenas letras.                     | Retorna `True` se só tiver letras e `False` caso contrário.  |

```python
# =============================================
# PASSO 5.4 — Manipulação
# =============================================

paciente = 'ANdre   '

print('Antes de tirar os espaços:',paciente)

paciente = paciente.strip()     # Remove os espaços
print('Depois:',paciente)

print("É uma string de apenas dígitos?",paciente.isdigit()) # Retorna True ou False se a string só contém NÚMEROS
print("É uma string de apenas letras?",paciente.isalpha())  # Retorna True ou False se a string só contém LETRAS

paciente_em_maiusculas = paciente.upper()   # Transforma todas as letras da string em maiúsculas
print("Nome em maiúsculas",paciente_em_maiusculas)

paciente_em_minusculas = paciente.lower()   # Transforma todas as letras da string em minúscula
print("Nome em minúsculas",paciente_em_minusculas)

Nome_paciente = paciente.capitalize()       # Deixa apenas a primeira letra da string maiúscula, e o resto minúscula
print("Nome com apenas a primeira letra maiúscula",Nome_paciente)


```

> 📝 Explicação do código acima:
> - `paciente.strip()` → Remove todos os espaços em branco (ou caracteres especiais de espaçamento) que existam antes ou depois do texto, útil para tratar entradas de dados "sujas".
> - `paciente.isdigit()` e `paciente.isalpha()` → Métodos que retornam um booleano (`True` ou `False`). O `.isdigit()` verifica se a *string* contém apenas números, enquanto o `.isalpha()` verifica se contém apenas letras.
> - `paciente.upper()` e `paciente.lower()` → Transformam todo o conteúdo da *string* para letras maiúsculas ou minúsculas, respectivamente.
> - `paciente.capitalize()` → Garante que a primeira letra da *string* seja maiúscula e todas as demais sejam minúsculas.
