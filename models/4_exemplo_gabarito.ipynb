> ## **Problema 1 - Somando números**
Você ta criando um programa pra somar números, e quer somar os números **1** e **3**

### **Problema 1.1**
Faça um programa que some os números que você quer

```python
# Objetivo: somar dois números
"""
Usamos a função print() juntamente com o operador de adição (+) para somar os valores diretamente.
"""
print(1 + 3)
```

### **Problema 1.2**
Faça um programa que some dois números que você der

```python
# Objetivo: receber e somar dois números. Use o método input() para receber valores
"""
A função input() recebe o valor como texto (string). 
Precisamos usar int() para converter esses textos em números inteiros antes de somá-los.
"""
n1 = int(input("Digite o primeiro número: "))
n2 = int(input("Digite o segundo número: "))
print(n1 + n2)
```

### **Problema 1.3**
Faça um programa que calcule a diferença entre 10 e 5

```python
# Objetivo: subtrair valores de forma direta
"""
Assim como a soma, utilizamos o operador de subtração (-) dentro do print().
"""
print(10 - 5)
```

### **Problema 1.4**
Faça um programa que receba três números do usuário e retorne a soma deles

```python
# Objetivo: receber múltiplos valores via input e somar
"""
Basta repetir a lógica do problema 1.2, adicionando uma terceira variável.
"""
a = int(input("Número 1: "))
b = int(input("Número 2: "))
c = int(input("Número 3: "))

print(a + b + c)
```

> ## **Problema 2 - Multiplicação e Divisão**
Praticando como multiplicar valores e fazer divisões justas.

### **Problema 2.1**
Faça um programa que multiplique os números 4 e 5

```python
# Objetivo: realizar uma multiplicação simples
"""
O operador de multiplicação no Python é o asterisco (*).
"""
print(4 * 5)
```

### **Problema 2.2**
Crie um programa que divida 20 por 4

```python
# Objetivo: realizar uma divisão exata
"""
O operador de divisão é a barra (/). Lembre-se que o resultado de uma divisão padrão sempre retorna um número float (com casas decimais).
"""
print(20 / 4)
```

### **Problema 2.3**
Faça um programa que receba dois valores e mostre o resto da divisão do primeiro pelo segundo

```python
# Objetivo: utilizar o operador de módulo (%)
"""
O operador '%' (módulo) retorna o resto da divisão entre dois números inteiros.
"""
n1 = int(input("Dividendo: "))
n2 = int(input("Divisor: "))
print(n1 % n2)
```

### **Problema 2.4**
Faça um programa que calcule a média entre três notas recebidas pelo usuário

```python
# Objetivo: somar valores e dividir pela quantidade
"""
Como são notas, podemos receber os valores como 'float' (números decimais).
Somamos tudo primeiro (usando parênteses para respeitar a ordem de precedência) e depois dividimos por 3.
"""
nota1 = float(input("Nota 1: "))
nota2 = float(input("Nota 2: "))
nota3 = float(input("Nota 3: "))

media = (nota1 + nota2 + nota3) / 3
print(media)
```

> ## **Problema 4 - Lidando com Listas**
Agrupando múltiplos valores em uma única variável.

### **Problema 4.1**
Crie uma lista contendo os números 1, 2 e 3 e a imprima

```python
# Objetivo: declarar e exibir uma lista
"""
Listas são criadas usando colchetes [ ] e os valores são separados por vírgula.
"""
numeros = [1, 2, 3]
print(numeros)
```

### **Problema 4.2**
Crie uma lista vazia, adicione o número 10 e imprima a lista

```python
# Objetivo: utilizar o método .append()
"""
O método .append() insere um novo item no final de uma lista existente.
"""
lista = []
lista.append(10)
print(lista)
```

### **Problema 4.3**
Dada a lista `frutas = ['maçã', 'banana', 'uva']`, imprima apenas o primeiro item

```python
# Objetivo: acessar elementos usando o índice
"""
No Python, o primeiro elemento de uma lista está no índice 0.
"""
frutas = ['maçã', 'banana', 'uva']
print(frutas[0])
```

### **Problema 4.4**
A partir da mesma lista de frutas, substitua 'banana' por 'laranja' e mostre a nova lista

```python
# Objetivo: modificar elementos existentes em uma lista
"""
Identificamos o índice do item (banana está no índice 1) e atribuímos o novo valor a ele.
"""
frutas = ['maçã', 'banana', 'uva']
frutas[1] = 'laranja'
print(frutas)
```

> ## **Problema 5 - Tomada de Decisão (If/Else)**
Permitindo que o programa faça escolhas baseadas em condições lógicas.

### **Problema 5.1**
Receba um número e verifique se ele é maior do que 10. Se for, imprima 'Maior que 10'

```python
# Objetivo: usar a estrutura condicional if simples
"""
Utilizamos 'if' seguido da condição. Lembre-se da indentação (espaço antes do print) para que o código pertença àquele bloco.
"""
n = int(input("Digite um número: "))
if n > 10:
    print("Maior que 10")
```

### **Problema 5.2**
Receba um número e informe se ele é positivo ou negativo (considere 0 como positivo)

```python
# Objetivo: usar if e else juntos
"""
O 'else' captura qualquer situação que não satisfaça a condição definida no 'if'.
"""
n = int(input("Digite um número: "))
if n >= 0:
    print("Positivo")
else:
    print("Negativo")
```

### **Problema 5.3**
Faça um programa que diga se um número recebido é par ou ímpar

```python
# Objetivo: juntar condicional if com o operador módulo (%)
"""
Um número par é aquele cujo resto da divisão por 2 é igual a zero.
Utilizamos == para verificar igualdade.
"""
n = int(input("Digite um número: "))
if n % 2 == 0:
    print("Par")
else:
    print("Ímpar")
```

### **Problema 5.4**
Receba dois números e imprima o maior entre eles. Se forem iguais, imprima 'Iguais'

```python
# Objetivo: usar if, elif e else para mais de duas condições
"""
O 'elif' nos permite adicionar condições extras quando precisamos checar mais de uma possibilidade específica.
"""
n1 = int(input("Número 1: "))
n2 = int(input("Número 2: "))

if n1 > n2:
    print(f"O maior é: {n1}")
elif n2 > n1:
    print(f"O maior é: {n2}")
else:
    print("Iguais")
```
