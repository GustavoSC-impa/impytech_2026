> ## **Problema 5 - Tomada de Decisão (If/Else)** 🟡
Permitindo que o programa faça escolhas baseadas em condições lógicas.

### ✅ **Resolução — Problema 5.1**
Um brinquedo no parque de diversões só permite a entrada de crianças com mais de 10 anos.

**🎯 Objetivo:** Utilizar a estrutura condicional simples para verificar uma condição.<br>

**🧠 Explicação lógica:**<br>
Utilizamos 'if' seguido da condição lógica. Lembre-se de respeitar a indentação obrigatória do bloco Python.

```python
"""
Utilizamos 'if' seguido da condição. Lembre-se da indentação (espaço antes do print) para que o código pertença àquele bloco.
"""
n = int(input("Digite um número: "))
if n > 10:
    print("Maior que 10")
```

### ✅ **Resolução — Problema 5.2**
Um aplicativo financeiro precisa classificar o balanço da sua conta do mês atual.

**🎯 Objetivo:** Utilizar a estrutura condicional composta para validar se um número é positivo ou negativo.<br>

**🧠 Explicação lógica:**<br>
O termo 'else' captura qualquer situação residual que não tenha satisfeito a condição definida previamente pelo 'if'.

```python
"""
O 'else' captura qualquer situação que não satisfaça a condição definida no 'if'.
"""
n = int(input("Digite um número: "))
if n >= 0:
    print("Positivo")
else:
    print("Negativo")
```

### ✅ **Resolução — Problema 5.3**
Em um jogo de par ou ímpar, o computador precisa validar automaticamente a resposta lendo a mão dos jogadores.

**🎯 Objetivo:** Verificar a paridade de um número recebido do usuário.<br>

**🧠 Explicação lógica:**<br>
Um número par é aquele cujo resto da divisão por 2 resulta exatamente em zero. Validamos a igualdade usando `==`.

```python
"""
Um número par é aquele cujo resto da divisão por 2 é igual a zero.
Utilizamos == para verificar igualdade.
"""
n = int(input("Digite um número: "))
if n % 2 == 0:
    print("Par")
else:
    print("Ímpar")
```

### ✅ **Resolução — Problema 5.4**
Dois jogadores tiraram dados e você precisa determinar qual valor foi mais alto ou se houve empate.

**🎯 Objetivo:** Analisar múltiplas condições lógicas.<br>

**🧠 Explicação lógica:**<br>
A instrução 'elif' nos permite concatenar avaliações lógicas extras quando precisamos validar mais de duas ramificações exclusivas.

```python
"""
O 'elif' nos permite adicionar condições extras quando precisamos checar mais de uma possibilidade específica.
"""
n1 = int(input("Número 1: "))
n2 = int(input("Número 2: "))

if n1 > n2:
    print(f"O maior é: {n1}")
elif n2 > n1:
    print(f"O maior é: {n2}")
else:
    print("Iguais")
```

> ## **Problema 6 - Caixa Eletrônico** 🔴
Você foi contratado para programar o sistema de saque de um banco. O cliente digita o valor que deseja sacar (considere sempre um valor múltiplo de 10).

### ✅ **Resolução — Problema 6.1**
O seu programa deve calcular e exibir a menor quantidade de notas de R$50, R$20 e R$10 necessárias para entregar o dinheiro exigido.

**🎯 Objetivo:** Decompor todo o valor do saque em notas menores.<br>

**🧠 Explicação lógica:**<br>
Para obter a *menor* quantidade de notas, sempre começamos pela nota de maior valor. Usamos a divisão inteira (`//`) para saber quantas notas daquele valor cabem no total. Em seguida, usamos o resto da divisão (`%`) para atualizar o valor que ainda falta ser pago. Repetimos o processo para as notas menores.

```python
"""
Passo 1: Receber o valor desejado (lembre-se de converter para int)
"""
valor = int(input("Qual valor deseja sacar? R$ "))

"""
Passo 2: Calcular as notas de 50 e descobrir o que sobra
"""
notas_50 = valor // 50
resto = valor % 50

"""
Passo 3: Usar o que sobrou para calcular as notas de 20, e atualizar o resto novamente
"""
notas_20 = resto // 20
resto = resto % 20

"""
Passo 4: O que sobrar agora são as notas de 10
"""
notas_10 = resto // 10

"""
Passo 5: Exibir o resultado final
"""
print(f"Notas de 50: {notas_50}")
print(f"Notas de 20: {notas_20}")
print(f"Notas de 10: {notas_10}")
```
