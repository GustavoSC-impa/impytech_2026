> ## **Conceito 1 - O que são variáveis?**
Uma **variável** é um espaço na memória do computador onde guardamos um valor para usar depois. Pense nela como uma caixinha com um nome — você coloca algo dentro e chama pelo nome quando precisar.

> **O que vamos aprender:**
> - Como criar e nomear variáveis em Python
> - Como exibir o valor de uma variável com `print()`
> - Como atualizar o valor de uma variável

### **Passo 1.1 — Criando e exibindo variáveis**
Vamos criar nossas primeiras variáveis e ver como o Python armazena e exibe seus valores.

```python
# =============================================
# PASSO 1.1 — Criando e exibindo variáveis
# =============================================

# [PARTE 1] — Criando variáveis com diferentes valores
nome = "Ana"          # variável do tipo texto (str)
idade = 20            # variável do tipo inteiro (int)
altura = 1.65         # variável do tipo decimal (float)

# [PARTE 2] — Exibindo os valores com print()
print(nome)           # exibe: Ana
print(idade)          # exibe: 20
print(altura)         # exibe: 1.65

#  AGORA É SUA VEZ: crie uma variável chamada 'cidade' com o nome da sua cidade e imprima ela
```

> **Explicação do código acima:**
> - `nome = "Ana"` → o sinal `=` **atribui** o valor `"Ana"` à variável `nome`
> - Textos sempre ficam entre aspas (`"` ou `'`); números **não** usam aspas
> - `print()` → função que exibe um valor na tela
> - **Ponto de atenção:** Python diferencia maiúsculas de minúsculas — `Nome` e `nome` são variáveis diferentes!

### **Passo 1.2 — Atualizando o valor de uma variável**
Variáveis podem ter seu valor trocado a qualquer momento. O Python sempre usa o valor mais recente.

```python
# =============================================
# PASSO 1.2 — Atualizando variáveis
# =============================================

# [PARTE 1] — Definindo o valor inicial
pontuacao = 0
print("Pontuação inicial:", pontuacao)   # exibe: 0

# [PARTE 2] — Atualizando o valor
pontuacao = 10
print("Após primeira fase:", pontuacao)  # exibe: 10

# [PARTE 3] — Incrementando com base no valor atual
pontuacao = pontuacao + 5                # soma 5 à pontuação atual
print("Após bônus:", pontuacao)          # exibe: 15

# AGORA É SUA VEZ: crie uma variável 'vidas' com valor 3, subtraia 1 e imprima o resultado
```

> **Explicação do código acima:**
> - `pontuacao = pontuacao + 5` → o Python primeiro calcula o lado direito (`pontuacao + 5`), depois guarda o resultado de volta na variável
> - `print("texto:", variavel)` → você pode passar múltiplos valores ao `print()` separados por vírgula
> - **Ponto de atenção:** a ordem importa! Uma variável precisa existir antes de ser usada em uma expressão.

> ## **Conceito 2 - Tipos de Dados**
Cada variável em Python tem um **tipo**, que define que tipo de informação ela carrega e o que você pode fazer com ela. Os 4 tipos básicos são: `int`, `float`, `str` e `bool`.

> **O que vamos aprender:**
> - Identificar os 4 tipos básicos do Python
> - Usar a função `type()` para descobrir o tipo de um valor
> - Converter entre tipos com `int()`, `float()` e `str()`

### **Passo 2.1 — Conhecendo os tipos básicos**
Vamos ver um exemplo de cada tipo e aprender a usar `type()` para identificá-los.

```python
# =============================================
# PASSO 2.1 — Tipos básicos do Python
# =============================================

# [PARTE 1] — Um exemplo de cada tipo
numero_inteiro = 42           # int   → números sem casa decimal
numero_decimal = 3.14         # float → números com casa decimal
texto          = "Python"     # str   → sequência de caracteres
verdadeiro     = True         # bool  → apenas True ou False

# [PARTE 2] — Descobrindo o tipo com type()
print(type(numero_inteiro))   # exibe: <class 'int'>
print(type(numero_decimal))   # exibe: <class 'float'>
print(type(texto))            # exibe: <class 'str'>
print(type(verdadeiro))       # exibe: <class 'bool'>

#  AGORA É SUA VEZ: crie uma variável com seu ano de nascimento e verifique o tipo dela com type()
```

> **Explicação do código acima:**
> - `type(valor)` → retorna o tipo do valor entre parênteses
> - `True` e `False` são escritos com a **primeira letra maiúscula** em Python
> - **Ponto de atenção:** `42` (int) e `42.0` (float) parecem iguais, mas são tipos diferentes para o Python!

### **Passo 2.2 — Convertendo entre tipos**
Às vezes precisamos transformar um tipo em outro — por exemplo, converter o texto recebido pelo `input()` em número para fazer cálculos.

```python
# =============================================
# PASSO 2.2 — Conversão de tipos (casting)
# =============================================

# [PARTE 1] — O problema: input() sempre retorna str
valor_digitado = "10"          # simula o que o usuário digita
print(type(valor_digitado))    # exibe: <class 'str'>

# [PARTE 2] — Convertendo str → int para poder calcular
valor_inteiro = int(valor_digitado)
print(type(valor_inteiro))     # exibe: <class 'int'>
print(valor_inteiro + 5)       # exibe: 15  (agora funciona!)

# [PARTE 3] — Outros tipos de conversão
print(float("3.14"))           # str  → float : exibe 3.14
print(str(100))                # int  → str   : exibe '100'
print(int(9.99))               # float→ int   : exibe 9 (trunca!)

#  AGORA É SUA VEZ: converta a string "25" para float e some 0.5 a ela
```

> **Explicação do código acima:**
> - `int()`, `float()`, `str()` → funções de **conversão** (casting) de tipos
> - `int(9.99)` resulta em `9`, não `10` — ele **corta** a parte decimal, sem arredondar
> - **Ponto de atenção:** `int("abc")` causa erro! Só converta strings que realmente contenham números.

> ## **Conceito 3 - Entrada do Usuário com input()**
O `input()` permite que o programa **receba dados digitados** pelo usuário durante a execução. Todo valor recebido via `input()` chega como **string** — lembre-se de converter quando necessário!

> **O que vamos aprender:**
> - Usar `input()` para receber dados do usuário
> - Combinar `input()` com conversão de tipo
> - Montar mensagens personalizadas com f-strings

### **Passo 3.1 — Recebendo e usando dados do usuário**
Vamos criar um mini-programa interativo que recebe o nome e a idade do usuário.

```python
# =============================================
# PASSO 3.1 — Entrada de dados com input()
# =============================================

# [PARTE 1] — Recebendo o nome (str, não precisa converter)
nome = input("Qual é o seu nome? ")          # a mensagem aparece para o usuário

# [PARTE 2] — Recebendo a idade (precisa converter para int)
idade = int(input("Qual é a sua idade? "))   # converte direto na mesma linha

# [PARTE 3] — Exibindo uma mensagem personalizada com f-string
ano_nascimento = 2025 - idade
print(f"Olá, {nome}! Você nasceu por volta de {ano_nascimento}.")  
# f-string: coloque f antes das aspas e use {variavel} dentro do texto

# AGORA É SUA VEZ: peça também a cidade do usuário e inclua na mensagem de saída
```

> **Explicação do código acima:**
> - `input("mensagem")` → exibe a mensagem e espera o usuário digitar
> - `f"texto {variavel}"` → **f-string**: forma elegante de inserir variáveis dentro de um texto
> - **Ponto de atenção:** o espaço antes de fechar as aspas em `"Qual é o seu nome? "` é intencional — deixa a interface mais limpa para o usuário.
